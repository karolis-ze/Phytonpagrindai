```python
# Neapibreztas skaicius poziciniu argumentu: *args
def funkcija(*args):
    return f'{args}'

funkcija(1, 2, 3, 4, 5)
```

```python
def vidurkis(*args):
    return sum(args)/len(args)

a = vidurkis(8, 9, 10, 80, -5, -20, 70)
print(a)
print(type(a))
```

```python
# Neapibreztas skaicius raktiniu argumentu: **kwargs
def funkcija(**kwargs):
    return f'{kwargs}'

funkcija(a=1, b=2, c=3, d=4, e=5, f=6)
```

```python
"""
Parašykite programą, kuri sukurtų funkciją func(), kuri priimančią kintamą skaičių argumentų ir atspausdinančią kiekvieną jų reikšmę

Pastaba: sukurkite šią funkciją taip, kad ji galėtų priimti bet kokį skaičių argumentų, juos apdoroti ir parodyti kiekvieno argumento reikšmę.
"""
```

```python
def func(*args):
    for i in args:
        print(f'{i}')

func('Karo', 'Techin', 'Python', 2026, False)
```

```python
"""
Parašykite funkciją calculation(), kuri:
    priima du kintamuosius ir apskaičiuoja jų sudėtį bei atimtį
    funkcija turėtų grąžinti ir sumą, ir skirtumą vienu return sakiniu
"""
```

```python
def calculation(a, b):
    return (a + b), (a - b), (a * b), (a / b)

print(calculation(1, 2))
print(type(calculation(1, 2)))

suma, atimtis, daugyba, dalyba = calculation(1, 2)
print(f'Suma {suma + 40}')
print(f'Atimtis {atimtis}')
print(f'Daugyba {daugyba}')
print(f'Dalyba {dalyba}')
print(f'---')
suma, atimtis = calculation(1, 2)
print(f'Suma {suma}')
print(f'Atimtis {atimtis}')
```

```python
"""
Parašykite programą, kuri sukurtų funkciją show_employee() su šiomis specifikacijomis:
    Ji turėtų priimti darbuotojo vardą ir atlyginimą.
    Joje turėtų būti rodomi ir vardas, ir atlyginimas.
    Jei atlyginimas nenurodytas funkcijos iškvietime, numatytoji reikšmė turėtų būti 9000.
"""
```

```python
def show_employee(vardas, atlyginimas=9000):
    print(f'Vardas {vardas}')
    print(f'Atlyginimas {atlyginimas}')

show_employee('Karo', 1000)
show_employee('Techin')
```

```python
def funkcija(vardas):
    pavarde = 'Lina'
    return f'{vardas}{pavarde}'

print(funkcija('Karo'))
# print(vardas)  # NameError: name 'vardas' is not defined
# print(pavarde)  # NameError: name 'pavarde' is not defined
```

```python
globalus_argumentas = 'Hello'

def funkcija(vardas, globalus_argumentas):
    pavarde = 'Lina'
    return f'{globalus_argumentas} {vardas}{pavarde}'

print(funkcija('Karo', 'Goodbye'))
# print(vardas)  # NameError: name 'vardas' is not defined
# print(pavarde)  # NameError: name 'pavarde' is not defined
print(globalus_argumentas)
```
