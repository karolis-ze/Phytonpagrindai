```python
"""
Užduotis: „Išmanioji automobilių draudimo skaičiuoklė“
    Sukurkite programą, kuri padėtų nuspręsti, ar klientas gali gauti specialią nuolaidą automobilio draudimui ir kokia bus galutinė metinė įmoka.

    Reikalavimai programai:
        Kintamieji ir Input:
            Paprašykite vartotojo įvesti:
                Vairuotojo amžių (int).
                Vairavimo stažą metais (int).
                Ar automobilis yra sportinis (įvesti „taip“ arba „ne“).
                Ar vairuotojas per pastaruosius metus turėjo baudų (įvesti „taip“ arba „ne“).
        Bazinė kaina: Pradinė draudimo kaina visiems yra 500 EUR.
        Logika (naudojant if, else, and, or, not):
            Jaunas vairuotojas: Jei vairuotojas yra jaunesnis nei 25 metų arba turi mažesnį nei 2 metų stažą, prie kainos pridedama 200 EUR.
            Sportinis automobilis: Jei automobilis yra sportinis ir vairuotojas turi baudų, kaina padidėja dar 300 EUR.
            Speciali nuolaida: Vairuotojas gauna 100 EUR nuolaidą, jei jis atitinka visas šias sąlygas:
                Yra vyresnis nei 30 metų.
                Ir neturi baudų.
                Ir (turi daugiau nei 10 metų stažą arba jo automobilis nėra sportinis).
            Draudimo atsisakymas: Jei vairuotojas yra jaunesnis nei 18 metų, programa turi išvesti pranešimą, kad draudimas negalimas, ir toliau nieko neskaičiuoti.
        Išvestis: Naudodami f-string, pateikite galutinę kainą ir trumpą santrauką (pvz., „Vairuotojo amžius: 32, Stažas: 12. Galutinė kaina: 400 EUR“).
"""
```

```python
# Jūsų kodas
vairuotojo_amzius = int(input("Iveskite vairuotojo amziu:"))
vairuotojo_stazas = int(input("Iveskite vairuotojo staza metais (Pvz. 1, 2, 4)"))
automobilio_tipas = str.upper(input("Ar automobilis spoetinis: Taip/Ne"))
vairuotojas_baustas = str.upper(input("Ar vairuotojas turejo baudu: Taip/Ne"))

bazine_kaina = 500
if vairuotojo_amzius > 25 and vairuotojo_stazas < 2:
    bazine_kaina += 200
if automobilio_tipas == "TAIP" and vairuotojas_baustas == "TAIP":
    print("Vairuotojas turi baudu, todėl kaina padidės.")
    bazine_kaina += 300
if vairuotojo_amzius > 30 and vairuotojas_baustas == "NE" and (vairuotojo_stazas > 10 or automobilio_tipas == "NE"):
    print("Vairuotojas atitinka visas sąlygas, todėl gaus nuolaidą.")
    bazine_kaina -= 100
if vairuotojo_amzius < 18:
    print("Draudimas negalimas, nes vairuotojas jaunesnis nei 18 metų.")
print(f"Vairuotojo amžius: {vairuotojo_amzius}, Stažas: {vairuotojo_stazas}. Galutinė kaina: {bazine_kaina} EUR")


    
```

```python
"""
Užduotis: „Penktadienio dilema“
    Sukurkite programą, kuri pagal įvestus duomenis nusprendžia vakaro planą.

    Vartotojo įvestis (Input):
        Kiek pinigų sąskaitoje turi (skaičius).
        Ar lauke lyja? (taip/ne).
        Ar tavo geriausias draugas/draugė gali susitikti? (taip/ne).
        Kiek procentų telefono baterijos liko (skaičius).
    Logikos taisyklės (naudojant if, else, and, or, not):
        Jei telefono baterija yra mažiau nei 15% ir tu neturi kroviklio (tarkime, kroviklio neturi niekada), programa iškart turi pasakyti: „Lik namie, tėvai neprisiskambins!“ ir toliau nieko nebetikrinti.
    Planų parinkimas:
        1. Vakarėlis mieste: Galimas tik tada, jei turi daugiau nei 50 EUR biudžetą IR lauke nelyja.
        2. Kinas arba kavinė: Jei turi tarp 20 ir 50 EUR (imtinai) ARBA jei lauke lyja, bet tavo draugas gali susitikti.
        3. Pasivaikščiojimas parke: Jei lauke nelyja IR biudžetas yra mažesnis nei 20 EUR, bet draugas gali susitikti.
        4. Liūdnas scenarijus: Jei neatitinka nei viena iš viršuje paminėtų sąlygų (pvz., neturi pinigų ir draugas negali susitikti), programa turi išvesti: „Šį vakarą žiūrim Netflix vieni.“

    Naudodamas f-string, išveskite galutinį sprendimą ir primink, kiek pinigų liks (jei eisi į miestą, atimkite simbolinius 10 EUR „kelionei“).
"""
```

```python
# Jūsų kodas
pinigu_saskaitoje = float(input("Kiek pinigų sąskaitoje turi:"))
lauke_lyja = str.upper(input("Ar lauke lyja: Taip/Ne"))
draugas_gali_susitikti = str.upper(input("Ar tavo geriausias draugas gali susitikti: Taip/Ne"))
telefono_baterija = float(input("Kiek procentų telefono baterijos liko:"))  

if telefono_baterija < 15:
    print("Lik namie, tėvai neprisiskambins!")
elif pinigu_saskaitoje > 50 and lauke_lyja == "NE":
    pinigu_saskaitoje -= 10
    print(f"Eisi į vakarėlį mieste. Likę pinigai: {pinigu_saskaitoje} EUR")
elif (20 <= pinigu_saskaitoje <= 50) or (lauke_lyja == "TAIP" and draugas_gali_susitikti == "TAIP"):
    pinigu_saskaitoje -= 10
    print(f"Eisi į kiną arba kavinę. Likę pinigai: {pinigu_saskaitoje} EUR")
elif lauke_lyja == "NE" and pinigu_saskaitoje < 20 and draugas_gali_susitikti == "TAIP":
    print(f"Eisi pasivaikščioti parke. Likę pinigai: {pinigu_saskaitoje} EUR")
else:
    print("Šį vakarą žiūrim Netflix vieni.")
```

```python
"""
Užduotis: „Namų saugumo valdiklis“
    Sukurkite programą, kuri imituoja išmanaus namo smegenis. Programa turi nuspręsti, ar įjungti signalizaciją, ar įleisti svečią, ar tiesiog nieko nedaryti.

    1. Duomenys (Žodynas / Dictionary):
    Sukurkite žodyną rezimai, kuriame saugomi galimi namų būsenų pavadinimai (pvz., 1: "Namie", 2: "Išvykęs", 3: "Miegas").
    2. Vartotojo įvestis (Input / Menu):
        Išveskite meniu iš žodyno ir paprašykite vartotojo įvesti skaičių (režimą).
        Paprašykite įvesti:
            Ar aptiktas judesys? (taip/ne)
            Ar turite durų kodą? (taip/ne)
            Ar dabar yra nakties metas? (taip/ne)
    3. Logikos taisyklės (if, and, or, not):
        Programos sprendimas priklauso nuo pasirinkto režimo:
            REŽIMAS „Namie“: * Jei aptiktas judesys IR nėra durų kodo, programa turi rašyti: „Pranešimas: Kažkas vaikšto prie durų, bet durys užrakintos.“
                Jei turimas kodas, rašykite: „Sveiki sugrįžę!“
            REŽIMAS „Išvykęs“: * Jei aptiktas judesys ARBA neteisingas durų kodas, rašykite: „DĖMESIO: Įjungiama sirena! Įsilaužimas!“
                Tik jei NĖRA judesio IR NĖRA kodo bandymo, rašykite: „Namuose ramu.“
            REŽIMAS „Miegas“: * Jei aptiktas judesys IR (yra nakties metas ARBA nėra kodo), rašykite:   „Įspėjimas: Užfiksuotas įtartinas judesys nakties metu!“
                Jei judesio nėra, rašykite: „Saldžių sapnų.“
            Jeigu rėžimo nėra - išvesti klaidą
    4. Išvestis:
        Naudokite f-strings, kad parodytum, koks režimas buvo pasirinktas (ištrauite reikšmę iš žodyno pagal vartotojo įvestą skaitmenį).
"""
```

```python
# Jūsų kodas
zodynas = {
    1: "Namie",
    2: "Išvykęs",
    3: "Miegas"
}

for key, value in zodynas.items():
    print(f"{key}: {value}")

rezimas = int(input("Pasirinkite režimą (1, 2, 3): "))
aptiktas_judesys = str.upper(input("Ar aptiktas judesys: Taip/Ne"))
durys_kodas = str.upper(input("Ar turite durų kodą: Taip/Ne"))
nakties_metas = str.upper(input("Ar dabar yra nakties metas: Taip/Ne")) 

if rezimas == 1:  # Namie
    if aptiktas_judesys == "TAIP" and durys_kodas == "NE":
        print("Pranešimas: Kažkas vaikšto prie durų, bet durys užrakintos.")
    elif durys_kodas == "TAIP":
        print("Sveiki sugrįžę!")
elif rezimas == 2:  # Išvykęs
    if aptiktas_judesys == "TAIP" or durys_kodas == "NE":
        print("DĖMESIO: Įjungiama sirena! Įsilaužimas!")
    else:
        print("Namuose ramu.")
elif rezimas == 3:  # Miegas
    if aptiktas_judesys == "TAIP" and (nakties_metas == "TAIP" or durys_kodas == "NE"):
        print("Įspėjimas: Užfiksuotas įtartinas judesys nakties metu!")
    else:
        print("Saldžių sapnų.")
else:
    print("Klaida: Neteisingas režimas.")   
```
