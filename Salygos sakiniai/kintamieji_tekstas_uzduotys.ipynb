```python
"""
I užduotis
    Sukurkite du kintamuosius: a = 15 ir b = 20
    Išspausdinkite šių palyginimų rezultatus:
        - Ar a lygus b?
        - Ar a nelygus b?
        - Ar a mažesnis už b?
        - Ar a didesnis arba lygus b?
"""
```

```python
# Jūsų kodas
a = 15
b = 20
print("a = ", a, "b =", b)
print("Ar lygus a b?", a == b)
print("Ar nelygus a b?", a != b)
print("Ar a mazesnis uz b?:", a < b)
print("Ar a didesnis uz b?:", a > b)
print("Ar a mazesnis arba lygus b?:", a <= b)
print("Ar a didesnis arba lygus b?:", a >= b)
```

```python
"""
III užduotis
    Sukurkite tris kintamuosius: x = 10, y = 20, z = 15
    Naudodami if/elif/else, nustatykite ir išspausdinkite, kuris skaičius yra didžiausias
"""
```

```python
# Jūsų kodas
x = 10
y = 20
z = 15
print("Skaičiai:", "x =", x, ", y =", y, ", z =", z)
if x > y and x > z:
    print("Didziausias skaicius yra x =", x)
elif y > x and y > z:
    print("Didziusias skaicius yra y =", y)
else:
    print("Didziausias skaicius yra z =", z)
```

```python
"""
VI užduotis
    Sukurkite kintamąjį tekstas = "PYTHON"
    Parašykite if/elif/else sąlygas:
        - Jei tekstas yra tik didžiosiomis - išspausdinkite: "Tekstas didžiosiomis"
        - Jei tekstas yra tik mažosiomis - išspausdinkite: "Tekstas mažosiomis"
        - Kitu atveju: "Mišrus tekstas"
"""
```

```python
# Jūsų kodas
tekstas = "PYTHON"
if tekstas.isupper():
    print("Tekstas didžiosiomis")
elif tekstas.islower():
    print("Tekstas mažosiomis")
else:
    print("Mišrus tekstas")
```

```python
"""
X užduotis
    Sukurkite kintamuosius:
        vardas = "jonas"
        amzius = 16
        turi_bilieta = True
    Parašykite if/elif/else sąlygas:
        1. Jei vardas NĖRA didžiosiomis IR amžius < 18 IR turi bilietą
           - išspausdinkite: "Nepilnametis su bilietu, prašome patikrinimo"
        2. Jei amžius >= 18 IR turi bilietą
           - išspausdinkite: "Galite įeiti"
        3. Jei neturi bilieto
           - išspausdinkite: "Reikalingas bilietas"
        4. Visais kitais atvejais
           - išspausdinkite: "Patikrinkite savo duomenis"
"""
```

```python
# Jūsų kodas
vardas = "jonas"
amzius = 16
turi_bilieta = True
if not vardas.isupper() and amzius < 18 and turi_bilieta:
    print("Nepilnametis su bilietu, prašome patikrinimo")
elif amzius >= 18 and turi_bilieta:
    print("Galite įeiti")
elif not turi_bilieta:
    print("Reikalingas bilietas")
else:
    print("Patikrinkite savo duomenis")
```
