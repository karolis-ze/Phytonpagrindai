```python
"""
Sukurkite programą, kuri:
    Paprašo vartotojo įvesti:
        savo ūgį metrais (pvz., 1.75)
        savo svorį kilogramais (pvz., 68)
    Apskaičiuoja KMI (kūno masės indeksą)
        Naudodama if sąlygą, nustato svorio kategoriją:
            Jei KMI < 18.5 → „Per mažas svoris“
            Jei KMI ≥ 18.5 ir < 25 → „Normalus svoris“
            Jei KMI ≥ 25 ir < 30 → „Antsvoris“
            Jei KMI ≥ 30 → „Nutukimas“
    Išveda rezultatą naudodama f-string formatą. Atsakymą suapvalinkite iki dviejų skaičių po kablelio
"""
```

```python
# Jūsų kodas
vardas = input("Iveskite savo varda:")
ugis = float(input("Iveskite savo ugi metrais:"))
svoris = float(input("Iveskite savo svori kilogramais:"))
if ugis <= 0 or svoris <= 0:
    print("Ugis ir svoris turi buti teigiami skaiciai.")
else:
    kmi = svoris / (ugis ** 2)
    if kmi < 18.5:
        kategorija = "Per mažas svoris"
    elif 18.5 <= kmi < 25:
        kategorija = "Normalus svoris"
    elif 25 <= kmi < 30:
        kategorija = "Antsvoris"
    else:
        kategorija = "Nutukimas"
    print(f"{vardas}, jūsų KMI yra {kmi:.2f}. Svorio kategorija: {kategorija}.")
```

```python
"""
Sukurkite programą, kuri:
    Paprašo vartotojo įvesti:
        Egzamino balą (0–100)
        Projekto balą (0–100)
    Apskaičiuoja galutinį pažymį, kai galutinį pažymį su daro 70 proc. egzamino balas ir 30 proc. projekto balas
    Naudoti if sąlygą:
        Jei galutinis ≥ 50 → „Kursas išlaikytas“
        Jei galutinis < 50 → „Kursas neišlaikytas“
    Išveda rezultatą naudodama f-string formatą. Atsakymą suapvalinkite iki dviejų skaičių po kablelio
"""
```

```python
# Jūsų kodas
egzamino_balas = float(input("Iveskite egzamino bala (0-100):"))
projekto_balas = float(input("Iveskite projekto bala (0-100):"))
if not (0 <= egzamino_balas <= 100) or not (0 <= projekto_balas <= 100):
    print("Balai turi buti tarp 0 ir 100.")
else:
    galutinis = (egzamino_balas * 0.7) + (projekto_balas * 0.3)
    if galutinis >= 50:
        rezultatas = "Kursas išlaikytas"
    else:
        rezultatas = "Kursas neišlaikytas"
    print(f"Galutinis pažymys: {galutinis:.2f}. {rezultatas}.")
```

```python
"""
Sukurkite programą, kuri:
    Paprašo vartotojo įvesti:
        Per mėnesį sunaudotą elektros kiekį kilovatvalandėmis (kWh).
    Taikykite skirtingus tarifus:
        Jei sunaudota iki 100 kWh (imtinai) → kaina 0.15 € už kWh.
        Jei sunaudota daugiau nei 100 kWh →
            už pirmas 100 kWh taikoma 0.15 € kaina,
            už likusią dalį taikoma 0.22 € kaina.
    Naudodami if sąlygą apskaičiuokite galutinę sumą.
    Išveda rezultatą naudodama f-string formatą. Atsakymą suapvalinkite iki dviejų skaičių po kablelio

Jeigu sunaudota daugiau nei 100 kWh, negalite viso kiekio dauginti iš didesnio tarifo – pirmiausia reikia atskirti pirmas 100 kWh.
"""
```

```python
# Jūsų kodas
suvartotas_kWh = float(input("Iveskite suvartota elektros kieki kWh:"))
if suvartotas_kWh < 0:
    print("Suvartotas kiekis negali buti neigiamas.")
else:
    if suvartotas_kWh <= 100:
        kaina = suvartotas_kWh * 0.15
    else:
        kaina = (100 * 0.15) + ((suvartotas_kWh - 100) * 0.22)
    print(f"Galutinė suma už elektros energiją: {kaina:.2f} €.")
```

```python
"""
Sukurkite sąrašą (list) su 5 pažymiais.
    Pvz.: 8, 7, 9, 6, 10
Programa turi:
    Apskaičiuoti pažymių vidurkį.
    Naudodama if / else nustatyti:
        Jei vidurkis ≥ 8 → „Puikūs rezultatai“
        Jei vidurkis ≥ 5 ir < 8 → „Rezultatai vidutiniai“
        Jei vidurkis < 5 → „Rezultatai silpni“
    Rezultatą išveskite naudodami f-string.
"""
```

```python
# Jūsų kodas
pazymiai = [8, 7, 9, 6, 10]
vidurkis = sum(pazymiai) / len(pazymiai)
print(f"Vidurkis: {vidurkis:.2f}")
if vidurkis >=8:
    print("Puikus rezultatai!")
elif 5 <= vidurkis < 8:
    print("Rezultatai vidutiniai")
else:
    print("Rezultatai silpni")  

```

```python
"""
Sukurkite žodyną, kuriame būtų saugoma informacija apie pamokų skaičių skirtingomis savaitės dienomis:
    Pvz.:
        Pirmadienis – 6 pamokos
        Antradienis – 7 pamokos
        Trečiadienis – 5 pamokos
        Ketvirtadienis – 6 pamokos
        Penktadienis – 4 pamokos
    Paprašykite vartotojo įvesti savaitės dieną.
    Naudodami if / else:
        Jei tokia diena yra žodyne → parodykite, kiek pamokų tą dieną vyksta.
        Jei tokios dienos nėra → išveskite pranešimą, kad tokia diena nerasta.
    Rezultatą išveskite naudodami f-string
"""
```

```python
# Jūsų kodas
zodynas = {
    "Pirmadienis": 6,
    "Antradienis": 7,
    "Trečiadienis": 5,
    "Ketvirtadienis": 6,
    "Penktadienis": 4
}
savaites_diena = input("Iveskite savaites diena:")
if savaites_diena in zodynas:
    pamoku_skaicius = zodynas[savaites_diena]
    print(f"{savaites_diena} vyksta {pamoku_skaicius} pamokos.")
else:
    print(f"{savaites_diena} nerasta žodyne.")  
```

```python
"""
Sukurkite žodyną su transporto priemonėmis ir minimaliais amžiaus reikalavimais:
        transportas = {
            "Dviratis": 0,
            "Motoroleris": 14,
            "Automobilis": 18,
            "Sunkvežimis": 21
        }
    Paprašykite vartotojo įvesti savo:
        Amžių
        Ar turi vairuotojo pažymėjimą (taip / ne)
    Naudodami if/else:
        1. Jei žmogus turi vairuotojo pažymėjimą:
            gali naudoti transportą, kurį leidžia amžius
        2. Jei neturi pažymėjimo:
            gali naudoti tik „Dviratį“ arba „Motorolerį“

    Išveskite rezultatą naudojant f-string
"""
```

```python
# Jūsų kodas
zodynas = {
    "Dviratis": 0,
    "Motoroleris": 14,
    "Atomobilis": 18,
    "Sunkvežimis": 21    
}
amzius = int(input("Iveskite savo amziu:"))
vairuotojo_pazymejimas = input("Ar turite vairuotojo pazymejima (taip/ne):").strip().lower()
if vairuotojo_pazymejimas == "taip":
    galimi_transportai = [transportas for transportas, reikalavimas in zodynas.items() if amzius >= reikalavimas]
    if galimi_transportai:
        print(f"Galite naudoti šias transporto priemones: {', '.join(galimi_transportai)}.")
    else:
        print("Jums nėra leidžiama naudoti jokių transporto priemonių pagal jūsų amžių.")
elif vairuotojo_pazymejimas == "ne":
    galimi_transportai = [transportas for transportas, reikalavimas in zodynas.items() if amzius >= reikalavimas and transportas in ["Dviratis", "Motoroleris"]]
    if galimi_transportai:
        print(f"Galite naudoti šias transporto priemones: {', '.join(galimi_transportai)}.")
    else:
        print("Jums nėra leidžiama naudoti jokių transporto priemonių pagal jūsų amžių.")
```

```python
"""
Sukurkite programą, kuri veikia kaip paprastas skaičiuotuvas
    Programa turi parodyti meniu su operacijomis sudėti, atimti, dauginti, dalinti
    Paprašykite vartotojo įvesti:
        Veiksmą (1–4)
        Pirmą skaičių
        Antrą skaičių
    Naudodami if / elif / else apskaičiuokite rezultatą pagal pasirinkimą.
        Dalybos atveju tikrinkite, ar antras skaičius nėra 0.
    Rezultatą išveskite naudodami f-string.
"""
```

```python
# Jūsų kodas
veiksmas = input("Pasirinkite veiksmą (1 - Sudėti, 2 - Atimti, 3 - Dauginti, 4 - Dalinti): ")
pirmas_skaicius = float(input("Įveskite pirmą skaičių: "))
antras_skaicius = float(input("Įveskite antrą skaičių: "))
if veiksmas == "1":
    rezultatas = pirmas_skaicius + antras_skaicius
    print(f"Rezultatas: {rezultatas}")
elif veiksmas == "2":
    rezultatas = pirmas_skaicius - antras_skaicius
    print(f"Rezultatas: {rezultatas}")
elif veiksmas == "3":
    rezultatas = pirmas_skaicius * antras_skaicius
    print(f"Rezultatas: {rezultatas}")
elif veiksmas == "4":
    if antras_skaicius == 0:
        print("Dalyba iš nulio negalima.")
    else:
        rezultatas = pirmas_skaicius / antras_skaicius
        print(f"Rezultatas: {rezultatas}")
else:
    print("Neteisingas veiksmo pasirinkimas.")
```
