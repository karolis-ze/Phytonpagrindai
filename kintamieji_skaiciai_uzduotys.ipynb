```python
"""
III užduotis:
    Susikurkite tris kintamuosius skaičiams saugoti. Į juos įrašykite norimus skaičius.
    Raskite šių skaičių suma, skirtumą, sandaugą ir dalmenį.
    Atsakymus išveskite kartu su atliekamu veiksmu (pvz 8 + 2 + 4 = 14).
"""
```

```python
# Jūsų kodas
a = 20
b = 4
c = 2
print("Suma:", f"{a} + {b} + {c}", "=", a + b + c)
print("Skirtumas:", f"{a} - {b} - {c}", "=", a - b - c)
print("Sandauga:", f"{a} * {b} * {c}", "=", a * b * c)
print("Dalyba:", f"{a} / {b} / {c}", "=", a / b / c)
```

```python
"""
IV užduotis:
    Susikurkite du kintamuosius skaičiams saugoti. Į juos įrašykite norimus skaičius.
    Pirmąjį kintamąjį padidinkite 5-iais.
    Antrajį padidinkite 2 kartus.
    Išveskite visus atsakymus (pradines reikšmes ir pakeistas reikšmes).
"""
```

```python
# Jūsų kodas
a = 10
b = 6

print("Pradines reiksmes:", "a =", a, ", b =", b)

a += 5
b += 2

print("Pakeistos reiksmes:", "a =", a, "b =", b)

```

```python
"""
V užduotis:
    Priskirkite savo pasirinktus skaičius kintamiesiems
        Raskite šių skaičių sandaugą, pakelkite juos savo pasirinktu
        laipsniu;
        Ištraukite iš šių skaičių pasirinkto laipsnio šaknį
        Suraskite šių skaičių dalybos liekaną (dalinkite didesnį iš
        mažesnio)
"""
```

```python
# Jūsų kodas
a = 9
print("Pradinis skaicius a:", a)
b = 4
print("Pradinis skaicius b:", b)

print("Sandauga:", a * b)
print("a pakelta 2 laipsniu:", a ** 2)
print("b pakelta 3 laipsniu:", b ** 3)
print("a is saknies:", a ** 0.5)
print("b is saknies:", b ** 0.5)
print("Dalybos liekana a dalinti is b:", a % b)
print("Dalybos liekana b dalinti is a:", b % a)
```
