```python
"""
1 užduotis – Pasisveikinimas
    Sąlyga:
        Paprašykite vartotojo įvesti savo vardą
        Išveskite asmeninį pasisveikinimą, panaudojant įvestą vardą (naudokite f-string)
"""
```

```python
# Jūsų kodas
vardas = input("Įveskite savo vardą: ")
print(f"Sveiki, {vardas}!")
```

```python
"""
2 užduotis – Amžiaus tikrinimas
    Sąlyga:
        Paprašykite vartotojo įvesti gimimo metus
        Apskaičiuokite, kiek jam metų 2026 metais
        Išveskite rezultatą sakiniu (naudokite f-string)
"""
```

```python
# Jūsų kodas
gimimo_metai = int(input("Įveskite savo gimimo metus: "))
amzius_2026 = 2026 - gimimo_metai
print(f"Jums bus {amzius_2026} metų 2026 metais)")
```

```python
"""
3 užduotis – Trikampio plotas su įvestimi
    Sąlyga:
        Paprašykite vartotojo įvesti trikampio pagrindą
        Paprašykite įvesti trikampio aukštinę
        Apskaičiuokite ir išveskite plotą (naudokite f-string)
"""
```

```python
# Jūsų kodas
trikampio_pagrindas = float(input("Įveskite trikampio pagrindą: "))
trikampio_aukstine = float(input("Įveskite trikampio aukštinę: "))
trikampio_plotas = (trikampio_pagrindas * trikampio_aukstine) / 2
print(f"Trikampio plotas yra: {trikampio_plotas}")
```

```python
"""
4 užduotis – Pažymių vidurkis
    Sąlyga:
        Paprašykite vartotojo įvesti tris pažymius (po vieną)
        Apskaičiuokite pažymių vidurkį
        Išveskite rezultatą (naudokite f-string)
        Rezultatą suapvalinkite iki dviejų skaičių (naudokite f-stringą arba .format())
"""
```

```python
# Jūsų kodas
pazymiai = []
for i in range(3):
    pazymys = float(input(f"Įveskite {i + 1}-ąjį pažymį: "))
    pazymiai.append(pazymys)
vidurkis = sum(pazymiai) / len(pazymiai)
print(f"Pažymių vidurkis yra: {vidurkis:.2f}")  
```

```python
"""
5 užduotis – Temperatūros konvertavimas
    Sąlyga:
        Paprašykite vartotojo įvesti temperatūrą Celsijaus laipsniais
        Konvertuokite ją į Farenheito laipsnius
        Išveskite abi temperatūras gražiai suformatuotame sakinyje (naudokite f-string)
"""
```

```python
# Jūsų kodas
temperatura_c = float(input("Įveskite temperatūrą Celsijaus laipsniais: "))
temperatura_f = (temperatura_c * 9/5) + 32
print(f"{temperatura_c}°C yra lygu {temperatura_f}°F")  
```

```python
"""
6 užduotis – Banko indėlis su palūkanomis
    Sąlyga:
        Paprašykite vartotojo įvesti pradinę indėlio sumą (eurais)
        Paprašykite įvesti metines palūkanas (procentais, pvz. 5 reiškia 5%)
        Paprašykite įvesti indėlio laikotarpį (metais)
        Apskaičiuokite galutinę sumą po nurodyto laikotarpio (su sudėtinėmis palūkanomis)
        Apskaičiuokite, kiek eurų uždirbo vartotojas
        Išveskite rezultatus (naudokite f-string, sumas apvalinkite iki dviejų skaičių naudodami f-stringą arba .format())

Formulė: Galutinė suma = Pradinė suma × (1 + palūkanos/100) ** metai
"""
```

```python
# Jūsų kodas
pradine_suma = float(input("Iveskite pradine indelio suma eurais:"))
metines_palukanos = int(input("Iveskite metines palukanas (procentais, pvz 5 reikskia 5%:"))
indelio_laikotarpis = int(input("Iveskite indelio laikotarpi (metais):"))
galutine_suma = pradine_suma * (1 + metines_palukanos/100) ** indelio_laikotarpis
print(f"Galutinė suma po {indelio_laikotarpis} metų bus: {galutine_suma:.2f} eurų")
uzdarbis = galutine_suma - pradine_suma
print(f"Uždirbta suma: {uzdarbis:.2f} eurų")
```

```python
"""
7 užduotis – Kelio skaičiuoklė su degalų kaina
    Sąlyga:
        Paprašykite vartotojo įvesti nukeliautą atstumą (km)
        Paprašykite įvesti automobilio vidutinę degalų sąnaudą (litrų/100km)
        Paprašykite įvesti degalų kainą už litrą (€)
        Apskaičiuokite:
            Kiek litrų degalų prireikė kelionei
            Kiek kainavo visa kelionė
            Kiek kainuoja 1 km nuvažiuoti
        Išveskite visus rezultatus (naudokite f-string)
"""
```

```python
# Jūsų kodas
nukeliautas_atstumas = int(input("Iveskite nukeliauta atstuma:"))
degalu_sanaudos = float(input("Iveskite degalu suvartojima (Litru/100km)"))
degalu_kaina = float(input("Iveskite degalu kaina (eurais):"))
degalu_kiekis = nukeliautas_atstumas * degalu_sanaudos / 100
keliones_kaina = degalu_kiekis * degalu_kaina
vienas_kilometras = keliones_kaina / nukeliautas_atstumas
print(f"Kelionei prireike {degalu_kiekis:.2f} degalu,")
print(f"Visa kelione kainavo {keliones_kaina:.2f} eurų")
print(f"Vienas 1km nuvaziuoti kainuoja: {vienas_kilometras:.2f} eurų")
```

```python
"""
8 užduotis – Laiko konvertavimas ir skaičiavimai
    Sąlyga:
        Paprašykite vartotojo įvesti valandas
        Paprašykite įvesti minutes
        Paprašykite įvesti sekundes
        Apskaičiuokite:
            Kiek iš viso sekundžių
            Kiek iš viso minučių (su tikslumu iki 2 ženklų)
            Kiek procentų paros sudaro šis laikas
        Papildomai: apskaičiuokite, kiek laiko liks iki paros pabaigos (24 val)
        Išveskite visus rezultatus (naudokite f-string)
"""
```

```python
# Jūsų kodas
valandos = int(input("Įveskite valandas: "))
minutes = int(input("Įveskite minutes: "))
sekundes = int(input("Įveskite sekundes: "))
viso_sekundziu = valandos * 3600 + minutes * 60 + sekundes
total_seconds = viso_sekundziu
total_minutes = total_seconds / 60
dienos_procentas = (total_seconds / 86400) * 100
liko_sekindziu = 86400 - total_seconds
liko_valandu = liko_sekindziu // 3600
liko_minuciu = (liko_sekindziu % 3600) // 60
visos_likusios_sekundes = liko_sekindziu % 60    
print(f"Viso sekundžių: {viso_sekundziu}")   
print(f"Viso minučių: {total_minutes:.2f}")
print(f"Šis laikas sudaro {dienos_procentas:.2f}% paros")
print(f"Liko iki paros pabaigos: {liko_valandu} valandų, {liko_minuciu} minučių, {visos_likusios_sekundes} sekundžių")

```
