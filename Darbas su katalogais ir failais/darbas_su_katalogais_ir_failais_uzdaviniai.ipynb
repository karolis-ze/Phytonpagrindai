```python
"""
I užduotis

Sukurti programą, kuri:
    Sukurtų failą „Tekstas.txt“ su pasinaudodami žemiau pateikta ištrauka iš Donelaitis - Metai
    Atspausdintų tekstą iš sukurto failo
    Paskutinėje sukurto failo eilutėje pridėtų šiandienos datą ir laiką
    Sunumeruotų teksto eilutes (kiekvienos pradžioje pridėtų skaičių).
    Sukurtame faile eilutę 'Jau saulelė vėl atkopdama budino svietą' pakeistų į 'Ale kokie dyvai – nei viens iš didelio pulko'
    Atspausdintų visą failo tekstą atbulai
    Nukopijuotų visą sukurto failą tekstą į naują failą, tik DIDŽIOSIOMIS RAIDĖMIS

Patarimai:
    Naudoti from datetime import datetime, datetime.today()
    Kintamajam priskirti sakinį, kuriuo bus operuojama, eilutėmis
"""
```

```python
eilerastis = """
Jau saulelė vėl atkopdama budino svietą
Ir žiemos šaltos trūsus pargraudama juokės.
Šalčių pramonės su ledais sugaišti pagavo,
Ir putodams sniegs visur į nieką pavirto.
Tuo laukus orai drungni gaivydami glostė
Ir žoleles visokias iš numirusių šaukė.
Krūmai su šilais visais išsibudino keltis,
O laukų kalnai su kloniais pametė skrandas.
Vislab, kas rudens bjaurybėj numirė verkdams,
Vislab, kas ežere gyvendams peržiemavojo
Ar po savo keru per žiemą buvo miegojęs,
Vislab tuo pulkais išlindo vasarą sveikyt.
Žiurkės su šeškais iš šalto pašalio traukės.
Varnos ir varnai su šarkoms irgi pelėdoms,
Pelės su vaikais ir kurmiai šilumą gyrė.
Musės ir vabalai, uodai su kaimene blusų
Mus jau vargyt vėl pulkais visur susirinko
Ir ponus taip, kaip būrus, įgelt išsižiojo.
Bet ir bitins jau šeimyną savo pabudyt
Ir prie darbo siųst bei ką pelnyt n’užsimiršo.
Tuo pulkai jų pro plyšius išlįsti pagavo
Ir lakstydami su birbynėms žaisti pradėjo;
O vorai kampuos sėdėdami verpalus audė
Irgi medžiot tinklus tyloms kopinėdami mezgė.
Bet ir meškos, ir vilkai šokinėdami džiaugės
Ir suplėšyt ką tyloms į pagirį traukės.
"""
```

```python
# Jūsų kodas
with open('Tekstas.txt', 'w') as failas:
    failas.write(eilerastis.strip())    

with open('Tekstas.txt', 'r') as failas:
    print(failas.read())
```

```python
from datetime import date
from datetime import datetime
with open ('Tekstas.txt', 'a', encoding='utf-8') as failas:
    with open('Tekstas.txt', 'a', encoding='utf-8') as failas:
        data = datetime.today()
        failas.write(f"\nŠiandien yra {data:%Y %m %d, %H:%M}\n")

    with open('Tekstas.txt', 'r', encoding='utf-8') as failas:
        print(failas.read())
    failas.write(data)

with open('Tekstas.txt', 'r') as failas:
    print(failas.read())
```

```python
for i, eilute in enumerate(eilerastis.splitlines(), start=0):
    print(f"{i}: {eilute}") 
```

```python
with open('Tekstas.txt', 'r') as failas:
    tekstas = failas.read()
    with open('Tekstas.txt', 'r', encoding='utf-8') as failas:
        tekstas = failas.read()

    tekstas = tekstas.replace(
        'Jau saulelė vėl atkopdama budino svietą',
        'Ale kokie dyvai – nei viens iš didelio pulko'
    )

    with open('Tekstas.txt', 'w', encoding='utf-8') as failas:
        failas.write(tekstas)
```
