```python
def skaiciu_suma(*skaiciai):
    suma = 0
    for skaicius in skaiciai:
        suma += skaicius 
    return suma

def skaiciaus_saknis(skaicius):
    if skaicius < 0:
        return "Klaida neigiamas skaicius neturi saknies"
    return skaicius ** 0.5

def kiek_zodis_simboliu(zodis):
    return len(zodis)

def dalyba(x, y):
    return x / y

suma = skaiciu_suma(5,6)
print(suma)

saknis = skaiciaus_saknis(16)
print(saknis)

zodis = kiek_zodis_simboliu("vasara")
print(zodis)

dalinti = dalyba(25,5)
print(dalinti)
```
