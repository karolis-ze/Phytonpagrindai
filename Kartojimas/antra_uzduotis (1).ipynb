## 1 dalis. Baudų skaičiavimas (`if` sąlygos)

Parašykite funkciją `bauda(dienos)`, kuri **grąžina** baudą eurais už pavėluotai grąžintą knygą:

| Vėlavimas | Bauda |
| --- | --- |
| 0 dienų | 0 EUR |
| 1–7 dienos | 0.50 EUR už kiekvieną dieną |
| daugiau nei 7 dienos | 3.50 EUR už pirmas 7 dienas + 1.00 EUR už kiekvieną papildomą dieną |
| **riba** | bendra bauda niekada neviršija **20 EUR** |

````python
"""
0 d.  ->  0.00 EUR
3 d.  ->  1.50 EUR
7 d.  ->  3.50 EUR
14 d. -> 10.50 EUR
30 d. -> 20.00 EUR (riba)
"""
````

````python
# Jūsų kodas

riba = 20
def bauda(dienos):
    if dienos < 0:
        return "minus negali buti"
    elif dienos == 0:
        skaiciuoti_bauda = 0.0
    elif 1 <= dienos <= 7:
        skaiciuoti_bauda = dienos * 0.5
    else:
        skaiciuoti_bauda = 3.5 + (dienos - 7) * 1.0
    if skaiciuoti_bauda > 20:
        return 20.0

    return skaiciuoti_bauda

for rinkinys in skolos:
    skaitytojas, knyga, dienos = rinkinys
    gauta_bauda = bauda(dienos)
    print(f"{dienos} d. --> {gauta_bauda:.2f} EUR")

````

## 2 dalis. Ataskaita (`for` ciklas)

**a)** Parašykite funkciją `skolos_tipas(dienos)`, kuri grąžina tekstą:
`'laiku'`, `'nedidelė'`, `'didelė'` arba `'labai didelė'`.

| Vėlavimas | Tipas |
| --- | --- |
| 0 dienų | `'laiku'` |
| 1–7 dienos | `'nedidelė'` |
| 8–14 dienų | `'didelė'` |
| daugiau nei 14 dienų | `'labai didelė'` |

**b)** Parašykite funkciją `ataskaita(skolos)`, kuri `for` ciklu su `enumerate()` išveda po
vieną eilutę kiekvienam skolininkui:

```
Nr. 4 | Aistė    | Sniego karalienė |  14 d. |  10.50 EUR | didelė
```

ir grąžia **du** kiekius: bendrą baudų sumą ir „labai didelių“ skolų skaičių.

````python
"""
Nr. 1 | Jonas    | Alchemikas       |   0 d. |   0.00 EUR | laiku
Nr. 2 | Rūta     | Hobitas          |   3 d. |   1.50 EUR | nedidelė
Bendra suma: 1.50 EUR, labai didelių: 0
"""
````

````python
# Jūsų kodas

def skolos_tipas(dienos):
    if dienos < 0:
        return "Negali buti minusas"
    elif dienos == 0:
        return "laiku"
    elif 1 <= dienos <= 7:
        return "nedidele"
    elif 8 <= dienos <= 14:
        return "didele"
    else:
        return "labai didele"

#koks_velavimas = skolos_tipas(6)
# print(f"Skolos tipas: {koks_velavimas}")
def ataskaita(skolos):
    labai_didele = 0
    bendra_suma = 0
    for i, rinkinys in enumerate(skolos, start=1):
        skaitytojas, knyga, dienos = rinkinys
        koks_velavimas = skolos_tipas(dienos)
        gauta_bauda = bauda(dienos)
        bendra_suma += gauta_bauda
        if dienos > 14:
            labai_didele += 1
        print(f"Nr. {i} | {skaitytojas} | {knyga} | {dienos} d.| {gauta_bauda} EUR | {koks_velavimas}")
    print(f"Bendra suma: {bendra_suma} EUR, labai dideliu: {labai_didele} ")
    return bendra_suma, labai_didele

ataskaita(skolos)
````

## 3 dalis. Funkcinis programavimas su `def` funkcijomis

Naudodami `map`, `filter`, `sorted` ir `max` bei savo `def` funkcijas, parašykite:

**Pagalbinės funkcijos, kurių argumentas — viena skola:**

- **a)** `bauda_is_skolos(skola)` — grąžina tos skolos baudą;
- **b)** `ar_pavelavo(skola)` — grąžina `True`, jei knyga grąžinta pavėluotai.

**Funkcijos, kurių argumentas — visas sąrašas:**

- **c)** `baudu_sarasas(skolos)` — visų baudų sąrašas (`map`);
- **d)** `pavelavusios_skolos(skolos)` — tik pavėluotos skolos (`filter`);
- **e)** `pagal_bauda(skolos)` — skolos, surikiuotos pagal baudą **mažėjančiai**
  (`sorted` su `key=bauda_is_skolos` ir `reverse=True`);
- **f)** `didziausia_bauda(skolos)` — didžiausią baudą turinti skola (`max` su
  `key=bauda_is_skolos`);
- **g)** `baudu_suma(skolos)` — bendra baudų suma (`map` + `sum`).

````python
"""
Baudų sąrašas:      [0, 1.5, 5.5, 10.5, 3.5, 20]
Pavėlavo:           5 skolininkai
Pagal baudą:        [('Gabija', 'Mažasis princas', 30), ('Aistė', 'Sniego karalienė', 14), ('Petras', 'Dune', 9), ('Marius', '1984', 7), ('Rūta', 'Hobitas', 3), ('Jonas', 'Alchemikas', 0)]
Didžiausia bauda:   ('Gabija', 'Mažasis princas', 30)
Bendra suma:        41.00 EUR
"""
````

````python
# Jūsų kodas

def bauda_is_skolos(skola):
    skaitytojas, knyga, dienos = skola
    skolos_dydis = bauda(dienos)
    return skolos_dydis

def ar_pavelavo(skola):
    skaitytojas, knyga, dienos = skola
    if dienos > 0:
        return True
    else:
        return False

def baudu_sarasas(skolos):
    return list(map(bauda_is_skolos, skolos))

def pavelavusios_skolos(skolos):
    return list(filter(ar_pavelavo, skolos))

def pagal_bauda(skolos):
    return sorted(skolos, key=bauda_is_skolos, reverse=True)

def didziausia_bauda(skolos):
    return max(skolos, key=bauda_is_skolos)

def baudu_suma(skolos):
    return sum(map(bauda_is_skolos, skolos))

print(f"Baudu sarasas: {baudu_sarasas(skolos)}")
print(f"Pavelavo: {len(pavelavusios_skolos(skolos))} skolininkai")
print(f"Pagal bauda: {pagal_bauda(skolos)}")
print(f"Didziausia bauda: {didziausia_bauda(skolos)}")
print(f"Bendra suma: {baudu_suma(skolos)}")
````

## 4 dalis. `while` ciklas

Bauda auga kiekvieną dieną, todėl galime apskaičiuoti, **kada** ji pasieks nustatytą ribą.

Parašykite funkciją `dienos_iki_ribos(dienos, riba=20)`, kuri **while** ciklu suskaičiuotų,
po kiek papildomų dienų bauda pasieks `riba`:

1. pradedame nuo `laukta = 0`;
2. **kol** `bauda(dienos + laukta) < riba`, didiname `laukta` vienetu;
3. grąžiname, kiek dienų prireikė.

**Pavyzdys:** jei skola vėluoja 9 dienas (bauda 5.50 EUR), 20 EUR ribą ji pasieks po 15 dienų.


````python
"""
14 d. (bauda 10.50 EUR) -> 20 EUR po 10 d.
30 d. (bauda 20.00 EUR) -> 20 EUR po 0 d.
"""
````

````python
# Jūsų kodas


def dienos_iki_ribos(dienos, riba=20):
    laukta = 0
    while bauda(dienos + laukta) < riba:
        laukta += 1
    return laukta


for rinkinys in skolos:
    skaitytojas, knyga, dienos = rinkinys
    dabar_bauda = bauda(dienos)
    papildomai_dienu = dienos_iki_ribos(dienos)
    print(f"{skaitytojas} ({dienos} d. {dabar_bauda:.2f} EUR) --> 20 EUR riba pasieks po {papildomai_dienu}")
````

## 5 dalis. `main()` — visos programos paleidimas

Parašykite funkciją `main()`, kuri sujungia visas dalis į vieną programą:

1. atspausdina ataskaitos antraštę (pvz., `=` simbolių eilutę ir pavadinimą);

4. išveda didžiausią baudą turintį skolininką (vardas, knyga, bauda);
5. pavyzdžiui, parodo, po kiek dienų 9 dienas vėluojanti skola pasieks 20 EUR ribą.

````python
"""
Bendra baudų suma:   41.00 EUR
Labai didelių skolų: 1

Skolininkai pagal baudą (mažėjančiai):
  Gabija    20.00 EUR
  Aistė     10.50 EUR
  Petras     5.50 EUR
  Marius     3.50 EUR
  Rūta       1.50 EUR
  Jonas      0.00 EUR

Didžiausia bauda: Gabija (Mažasis princas) - 20.00 EUR
9 d. vėluojanti skola 20 EUR ribą pasieks po 15 d.
"""
````

````python
# Jūsų kodas


def main(skolos):
    print("=" * 30)
    print("BIBLIOTEKOS ATASKAITA")
    print("=" * 30)

    print(f"\nBendra baudu suma: {baudu_suma(skolos):.2f} EUR")
    dideles = filter(lambda s: bauda_is_skolos(s) >=20, skolos)
    print(f"Labai dideliu skolu {len({dideles})}")
    print("\nSkolininkai pagal bauda (mazejanciai)")

    sarasas = pagal_bauda(skolos)
    for rinkinys in sarasas:
        skaitytojas, knyga, dienos = rinkinys
        print(f"{skaitytojas:<10} {bauda(dienos):.2f} EUR ")

    dskaitytojas, dknyga, ddienos = didziausia_bauda(skolos)
    print(f"\nDidziausia bauda {dskaitytojas} ({dknyga}) - {bauda(ddienos)} EUR ")
    v_diena = dienos_iki_ribos(9)
    print(f"9 d. vėluojanti skola 20 EUR ribą pasieks po {v_diena} dienu.")
    

    
    
main(skolos)
````
