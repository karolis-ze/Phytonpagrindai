````python
# Jūsų kodas
lankytojai = [
    (7, False),
    (5, False),
    (16, False),
    (20, True),
    (35, False),
    (70, False),
    (68, True),
    (17, True),
]
kaina = []
def bilieto_kaina(amzius, ar_studentas):
    if amzius < 7:
        return 0
    elif amzius < 18:
        return 5
    elif amzius >= 65:
        return 6
    elif ar_studentas:
        return 4
    else:
        return 8

print("Bilieto kainos:")
for lankytojas in lankytojai:
    amzius, ar_studentas = lankytojas
    kaina.append(bilieto_kaina(amzius, ar_studentas))
    print(kaina[-1])
````

2. grąžia **du** kiekius: bendrą sumą ir nemokamų bilietų skaičių.

````python
"""
Nr. 1 |   7 m. | mokinys    |  5.00 EUR
Nr. 2 |   5 m. | nemokamai  |  0.00 EUR
Nr. 3 |  16 m. | mokinys    |  5.00 EUR
Nr. 4 |  20 m. | studentas  |  6.00 EUR
Nr. 5 |  35 m. | suaugęs    |  8.00 EUR
Nr. 6 |  70 m. | senjoras   |  4.00 EUR
Nr. 7 |  68 m. | senjoras   |  4.00 EUR
Nr. 8 |  17 m. | mokinys    |  5.00 EUR
---------------------------------------------
Bendra suma:      37.00 EUR
Nemokamų bilietų: 1
"""
````

````python
# Jūsų kodas
lankytojai = [
    (7, False),
    (5, False),
    (16, False),
    (20, True),
    (35, False),
    (70, False),
    (68, True),
    (17, True),
]

def lankytojo_tipas(amzius, ar_studentas):
    if amzius < 7:
        return "nemokamai"
    elif amzius < 18:
        return "mokinys"
    elif amzius >= 65:
        return "senjoras"
    elif ar_studentas:
        return "Studentas"
    else:
        return "Pilnametis"
bendra_kaina = []
for index, lankytojas in enumerate(lankytojai):
    amzius, ar_studentas = lankytojas
    tipas = lankytojo_tipas(amzius, ar_studentas)
    kaina = bilieto_kaina(amzius, ar_studentas)
    print(f"Nr. {index + 1} | {amzius} m. | {tipas} | {kaina:.2f} EUR")
    bendra_kaina.append(kaina)
print("-" * 45)
print(f"Bendra suma: {sum(bendra_kaina):.2f} EUR")
nemokamu_bilietu_skaicius = sum(1 for lankytojas in lankytojai if bilieto_kaina(lankytojas[0], lankytojas[1]) == 0)
print(f"Nemokamų bilietų: {nemokamu_bilietu_skaicius}")

 

````

## 3 dalis. Funkcinis programavimas su `def` funkcijomis

Naudodami `map`, `filter`, `sorted` ir `max` bei savo `def` funkcijas, parašykite:

**Pagalbinės funkcijos, kurių argumentas — vienas lankytojas:**

- **a)** `amzius(lankytojas)` — grąžina lankytojo amžių;
- **b)** `kaina_is_lankytojo(lankytojas)` — grąžina lankytojo bilieto kainą;
- **c)** `ar_studento_nuolaida(lankytojas)` — grąžina `True`, jei lankytojui **tikrai**
  taikoma studento nuolaida, t. y. jo bilietas kainuoja 6 EUR.

**Funkcijos, kurių argumentas — visas sąrašas:**

- **d)** `kainu_sarasas(lankytojai)` — visų bilietų kainų sąrašas (`map`);
- **e)** `studentai_su_nuolaida(lankytojai)` — atrinkti lankytojai (`filter`);
- **f)** `pagal_amziu(lankytojai)` — lankytojai, surikiuoti pagal amžių didėjančiai
  (`sorted` su `key=amzius`);
- **g)** `brangiausias_lankytojas(lankytojai)` — lankytojas, kurio bilietas brangiausias
  (`max` su `key=kaina_is_lankytojo`).

````python
"""
Kainų sąrašas:        [5, 0, 5, 6, 8, 4, 4, 5]
Bendra suma (sum):    37.00 EUR
Studento nuolaida:    [(20, True)]
Pagal amžių (sorted): [(5, False), (7, False), (16, False), (17, True), (20, True), (35, False), (68, True), (70, False)]
Brangiausias bilietas: (35, False) -> 8.00 EUR
"""
````

````python
# Jūsų kodas
lankytojai = [
    (7, False),
    (5, False),
    (16, False),
    (20, True),
    (35, False),
    (70, False),
    (68, True),
    (17, True),
]

def amzius(lankytojai):
    return sorted(lankytojai, key=lambda x: x[0])
def bilieto_kaina(amzius, ar_studentas):
    if amzius <= 6:
        return 0
    elif amzius < 18:
        return 5
    elif amzius >= 65:
        return 4
    elif ar_studentas:
        return 6
    else:
        return 8

def amzius(lankytojas):
    return lankytojas[0]


def kaina_is_lankytojo(lankytojas):
    return bilieto_kaina(*lankytojas)


def ar_studento_nuolaida(lankytojas):
    return kaina_is_lankytojo(lankytojas) == 6


def kainu_sarasas(lankytojai):
    return list(map(kaina_is_lankytojo, lankytojai))


def studentai_su_nuolaida(lankytojai):
    return list(filter(ar_studento_nuolaida, lankytojai))


def pagal_amziu(lankytojai):
    return sorted(lankytojai, key=amzius)


def brangiausias_lankytojas(lankytojai):
    return max(lankytojai, key=kaina_is_lankytojo)


print("Kainų sąrašas:       ", kainu_sarasas(lankytojai))
print("Bendra suma (sum):   ", f"{sum(kainu_sarasas(lankytojai)):.2f} EUR")
print("Studento nuolaida:   ", studentai_su_nuolaida(lankytojai))
print("Pagal amžių (sorted):", pagal_amziu(lankytojai))
brangiausias = brangiausias_lankytojas(lankytojai)
print("Brangiausias bilietas:", brangiausias, "->", f"{kaina_is_lankytojo(brangiausias):.2f} EUR")




````

## 4 dalis. Biudžetas (`while` ciklas)

Parašykite funkciją `sutalpinti_i_biudzeta(lankytojai, biudzetas)`, kuri:

1. **kol** bendra bilietų suma viršija biudžetą, iš sąrašo pašalina **brangiausią** bilietą
   (naudokite `while`);
2. grąžina likusių lankytojų sąrašą ir bendrą sumą.

**Pavyzdys:** biudžetas 20 EUR. Iš 37 EUR pašalinami 3 brangiausi bilietai — lieka
5 lankytojai už 18 EUR.

> **Dėmesio:** funkcija neturi pakeisti originalaus `lankytojai` sąrašo.

````python
"""
Biudžetas:         20 EUR
Telpa lankytojai:  [(5, False), (16, False), (70, False), (68, True), (17, True)]
Bendra suma:       18.00 EUR (5 bilietai)
"""
````

````python
# Jūsų kodas

lankytojai = [
    (7, False),
    (5, False),
    (16, False),
    (20, True),
    (35, False),
    (70, False),
    (68, True),
    (17, True),
]

def sutalpinti_i_biudzeta(lankytojai, biudzetas):
    while biudzetas > brangiausias_lankytojas (lankytojai):
        lankytojai.remove(brangiausias_lankytojas(lankytojai))  

print("Biudžetas:         20 EUR")
def sutalpinti_i_biudzeta(lankytojai, biudzetas):
    likę_lankytojai = lankytojai.copy()

    while sum(kainu_sarasas(likę_lankytojai)) > biudzetas:
        brangiausias = brangiausias_lankytojas(likę_lankytojai)
        likę_lankytojai.remove(brangiausias)

    return likę_lankytojai, sum(kainu_sarasas(likę_lankytojai))


print("Biudžetas:         20 EUR")
lankytojai, bendra_suma = sutalpinti_i_biudzeta(lankytojai, 20)
print("Telpa lankytojai:  ", pagal_amziu(lankytojai))
print("Bendra suma:       ", f"{bendra_suma:.2f} EUR", f"({len(lankytojai)} bilietai)")
print("Telpa lankytojai:  ", pagal_amziu(lankytojai))
print("Bendra suma:       ", f"{sum(kainu_sarasas(lankytojai)):.2f} EUR", f"({len(lankytojai)} bilietai)")
````
