## 1 dalis. Bilieto kainos funkcija

Parašykite funkciją `bilieto_kaina(amzius, ar_studentas)`, kuri **grąžina** bilieto kainą eurais:

| Lankytojas | Kaina |
| --- | --- |
| amžius iki 6 metų imtinai | 0 EUR (nemokamai) |
| amžius iki 18 metų | 5 EUR |
| senjoras (65 m. ir vyresnis) | 4 EUR |
| studentas (18 m. ir vyresnis) | 6 EUR |
| visi kiti | 8 EUR |

````python
"""
Pvz:
0
5
6
4
8
"""
````

````python
# Jūsų kodas
lankytojai = [
    (7, False),
    (5, False),
    (16, False),
    (20, True),
    (35, False),
    (70, False),
    (68, True),
    (17, True),
]
kaina = []
def bilieto_kaina(amzius, ar_studentas):
    if amzius <= 6:
        return 0
    elif amzius < 18:
        return 5
    elif amzius >= 65:
        return 4
    elif ar_studentas:
        return 6
    else:
        return 8

print("Bilieto kainos:")
for lankytojas in lankytojai:
    amzius, ar_studentas = lankytojas
    kaina.append(bilieto_kaina(amzius, ar_studentas))
    print(kaina[-1])
````

## 2 dalis. Kasos ataskaita (`for` ciklas)

**a)** Parašykite funkciją `lankytojo_tipas(amzius, ar_studentas)`, kuri grąžina tekstą:
`'nemokamai'`, `'mokinys'`, `'senjoras'`, `'studentas'` arba `'suaugęs'`.

**b)** Parašykite funkciją `kasa(lankytojai)`, kuri:

1. `for` ciklu su `enumerate()` išveda po vieną eilutę kiekvienam lankytojui:

```
Nr. 1 |   7 m. | mokinys    |   5.00 EUR
```

2. grąžia **du** kiekius: bendrą sumą ir nemokamų bilietų skaičių.

````python
"""
Nr. 1 |   7 m. | mokinys    |  5.00 EUR
Nr. 2 |   5 m. | nemokamai  |  0.00 EUR
Nr. 3 |  16 m. | mokinys    |  5.00 EUR
Nr. 4 |  20 m. | studentas  |  6.00 EUR
Nr. 5 |  35 m. | suaugęs    |  8.00 EUR
Nr. 6 |  70 m. | senjoras   |  4.00 EUR
Nr. 7 |  68 m. | senjoras   |  4.00 EUR
Nr. 8 |  17 m. | mokinys    |  5.00 EUR
---------------------------------------------
Bendra suma:      37.00 EUR
Nemokamų bilietų: 1
"""
````

````python
# Jūsų kodas
lankytojai = [
    (7, False),
    (5, False),
    (16, False),
    (20, True),
    (35, False),
    (70, False),
    (68, True),
    (17, True),
]

def lankytojo_tipas(amzius, ar_studentas):
    if amzius < 7:
        return "nemokamas"
    elif 7 <= amzius <= 18:
        return "mokinys"
    elif amzius >= 65:
        return "senjoras"
    elif ar_studentas:
        return "studentas"
    else:
        return "pilnametis"

def kasa(lankytojai):
    bendra_suma = 0
    nemokami_bilietai = 0

    for i, lankytojas in enumerate(lankytojai, start=1):
        amzius, ar_studentas = lankytojas
        tipas = lankytojo_tipas(amzius, ar_studentas)
        kaina = bilieto_kaina(amzius, ar_studentas)
    
        bendra_suma += kaina
        if kaina == 0:
            nemokami_bilietai += 1
        print(f"Nr. {i} | {amzius} m.| {tipas}| {kaina:.2f} EUR")
    return bendra_suma, nemokami_bilietai

bendra_suma, nemokami_bilietai = kasa(lankytojai)
print(f"Bendra bilietu suma: {bendra_suma}")
print(f"Nemokamu bilietu skaicius: {nemokami_bilietai}")


    

    
 

````

## 3 dalis. Funkcinis programavimas su `def` funkcijomis

Naudodami `map`, `filter`, `sorted` ir `max` bei savo `def` funkcijas, parašykite:

**Pagalbinės funkcijos, kurių argumentas — vienas lankytojas:**

- **a)** `amzius(lankytojas)` — grąžina lankytojo amžių;
- **b)** `kaina_is_lankytojo(lankytojas)` — grąžina lankytojo bilieto kainą;
- **c)** `ar_studento_nuolaida(lankytojas)` — grąžina `True`, jei lankytojui **tikrai**
  taikoma studento nuolaida, t. y. jo bilietas kainuoja 6 EUR.

**Funkcijos, kurių argumentas — visas sąrašas:**

- **d)** `kainu_sarasas(lankytojai)` — visų bilietų kainų sąrašas (`map`);
- **e)** `studentai_su_nuolaida(lankytojai)` — atrinkti lankytojai (`filter`);
- **f)** `pagal_amziu(lankytojai)` — lankytojai, surikiuoti pagal amžių didėjančiai
  (`sorted` su `key=amzius`);
- **g)** `brangiausias_lankytojas(lankytojai)` — lankytojas, kurio bilietas brangiausias
  (`max` su `key=kaina_is_lankytojo`).

````python
"""
Kainų sąrašas:        [5, 0, 5, 6, 8, 4, 4, 5]
Bendra suma (sum):    37.00 EUR
Studento nuolaida:    [(20, True)]
Pagal amžių (sorted): [(5, False), (7, False), (16, False), (17, True), (20, True), (35, False), (68, True), (70, False)]
Brangiausias bilietas: (35, False) -> 8.00 EUR
"""
````

````python
# Jūsų kodas
lankytojai = [
    (7, False),
    (5, False),
    (16, False),
    (20, True),
    (35, False),
    (70, False),
    (68, True),
    (17, True),
]

def amzius(lankytojas):
    return lankytojas[0]

def kaina_is_lankytojo(lankytojas):
    amzius, ar_studentas = lankytojas
    kaina = bilieto_kaina(amzius, ar_studentas)
    return kaina

def ar_studento_nuolaida(lankytojas):
    amzius, ar_studentas = lankytojas
    kaina = kaina_is_lankytojo(lankytojas)
    return kaina == 6 and ar_studentas

def kainu_sarasas(lankytojai):
    return list(map(kaina_is_lankytojo, lankytojai))

def studentai_su_nuolaida(lankytojai):
    return list(filter(ar_studento_nuolaida, lankytojai))

def pagal_amziu(lankytojai):
    return sorted(lankytojai, key=amzius)

def brangiausias_lankytojas(lankytojai):
    return max(lankytojai, key=kaina_is_lankytojo)

# 1. Kainu sarasas
kainos = kainu_sarasas(lankytojai)
print(f"Kainu sarasas: {kainos}")

# 2.Bendra suma
bendra_suma = sum(kainos)
print(f"Bendra suma: {bendra_suma:.2f} EUR")


tikri_studentai = studentai_su_nuolaida(lankytojai)
print(f"Studento nuolaida:{tikri_studentai}")

# 3. Pagal amziu
surikiuoti = pagal_amziu(lankytojai)
print(f"Pagal amziu (sorted): {surikiuoti}")

# 4. Brangiausias bilietas
brangiausias = brangiausias_lankytojas(lankytojai)
kaina = kaina_is_lankytojo(brangiausias)
print(f"Brangiausias bilietas {brangiausias} --- > {kaina:.2f} EUR")
````

## 4 dalis. Biudžetas (`while` ciklas)

Parašykite funkciją `sutalpinti_i_biudzeta(lankytojai, biudzetas)`, kuri:

1. **kol** bendra bilietų suma viršija biudžetą, iš sąrašo pašalina **brangiausią** bilietą
   (naudokite `while`);
2. grąžina likusių lankytojų sąrašą ir bendrą sumą.

**Pavyzdys:** biudžetas 20 EUR. Iš 37 EUR pašalinami 3 brangiausi bilietai — lieka
5 lankytojai už 18 EUR.

> **Dėmesio:** funkcija neturi pakeisti originalaus `lankytojai` sąrašo.

````python
"""
Biudžetas:         20 EUR
Telpa lankytojai:  [(5, False), (16, False), (70, False), (68, True), (17, True)]
Bendra suma:       18.00 EUR (5 bilietai)
"""
````

````python
# Jūsų kodas

lankytojai = [
    (7, False),
    (5, False),
    (16, False),
    (20, True),
    (35, False),
    (70, False),
    (68, True),
    (17, True),
]
biudzetas = 20

def sutalpinti_i_biudzeta(lankytojai, biudzetas):
    like_lankytojai = lankytojai.copy()
    bendra_suma = sum(kainu_sarasas(like_lankytojai))
    while bendra_suma > biudzetas:
        brangiausias = brangiausias_lankytojas(like_lankytojai)
        like_lankytojai.remove(brangiausias)
        bendra_suma = sum(kainu_sarasas(like_lankytojai))
    return like_lankytojai, bendra_suma

likusieji, bendra_suma = sutalpinti_i_biudzeta(lankytojai, biudzetas)

print(f"Biudzetas: {biudzetas} EUR")
print(f"Liko lankytoju: {len(likusieji)}")
print(f"Like lankytojai: {likusieji}")
print(f"Galutine suma: {bendra_suma} EUR")

````
