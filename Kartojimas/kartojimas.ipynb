# Kartojimas prieš egzaminą

Trumpa viso kurso teorija vienoje vietoje:

1. **Duomenų tipai** — ką saugo kintamasis ir ką su juo galima daryti;
2. **`if`** — kada reikia sąlygos;
3. **`for`** — kada kartojame žinomą skaičių kartų;
4. **`while`** — kada kartojame, kol sąlyga teisinga;
5. **`def`** — kada logiką išskiriame į funkciją.

## 1. Duomenų tipai

| Tipas | Pavyzdys | Ką saugo | Galima keisti vietoje? |
| --- | --- | --- | --- |
| `int` | `42` | sveikąjį skaičių | ne |
| `float` | `3.14` | dešimtainį skaičių | ne |
| `str` | `'Jonas'` | tekstą | ne |
| `bool` | `True`, `False` | tiesa / melas | ne |
| `list` | `[8, 9, 7]` | sąrašą (tvarka svarbi) | **taip** |
| `tuple` | `(54.68, 25.28)` | nekintantį sąrašą | ne |
| `set` | `{1, 2, 3}` | unikalias reikšmes (tvarka nesvarbi) | **taip** |
| `dict` | `{'vardas': 'Rūta'}` | poras raktas → reikšmė | **taip** |

**Svarbiausios operacijos:** `len()`, `sum()`, `min()`, `max()`, `sorted()`,
indeksas `x[0]`, paskutinis elementas `x[-1]`, išpjova `x[1:3]`, narystė `in`.

````python
# --- Skaičiai ---
skaicius = 42            # int
desimtainis = 3.14       # float
print(type(skaicius), skaicius)
print(type(desimtainis), desimtainis)
print(10 / 3, 10 // 3, 10 % 3, 2 ** 10)   # /, //, %, **

# --- Tekstas ---
vardas = 'Jonas'         # str
print(type(vardas), vardas, len(vardas))
print(vardas[0], vardas[-1], vardas[1:3])
print(vardas.upper(), vardas.lower())

# --- Loginės reikšmės ---
print(type(True), 10 > 3, 10 == 3, not False)   # bool

# --- Sąrašas ---
pazymiai = [8, 9, 7, 10]                        # list
print(type(pazymiai), pazymiai, len(pazymiai))
print(pazymiai[0], pazymiai[-1], pazymiai[1:3])
print(sum(pazymiai), max(pazymiai), min(pazymiai))

# --- Tuple (nekintantis sąrašas) ---
koordinates = (54.68, 25.28)                    # tuple
print(type(koordinates), koordinates, koordinates[0])

# --- Aibė (tik unikalios reikšmės) ---
unikalus = {3, 1, 2, 3, 1}                      # set
print(type(unikalus), unikalus)

# --- Žodynas (raktas -> reikšmė) ---
studentas = {'vardas': 'Rūta', 'amzius': 20}    # dict
print(type(studentas), studentas['vardas'], studentas['amzius'])
print(studentas.keys())
print(studentas.values())
````

### Ką rinktis?

| Situacija | Tipas |
| --- | --- |
| vienas skaičius | `int` arba `float` |
| tekstas | `str` |
| atsakymas „taip / ne“ | `bool` |
| keli kintantys duomenys (pažymiai, prekės) | `list` |
| fiksuotas rinkinys (koordinatės, data) | `tuple` |
| reikia unikalių reikšmių ir greitos `in` patikros | `set` |
| reikšmė pagal raktą (vardas → amžius) | `dict` |

**Kintami tipai** (`list`, `dict`, `set`) keičiasi vietoje — funkcija, kuri juos keičia,
pakeičia ir originalą. **Nekintami** (`int`, `float`, `str`, `bool`, `tuple`) lieka tokie patys;
norėdami „pakeisti“, privalote priskirti naują reikšmę.

````python
# --- list: galima keisti vietoje ---
pazymiai = [8, 9, 7]
pazymiai.append(10)
pazymiai[0] = 6
print(pazymiai)

# --- dict: galima keisti vietoje ---
studentas = {'vardas': 'Rūta', 'amzius': 20}
studentas['amzius'] = 21
studentas['miestas'] = 'Vilnius'
print(studentas)

# --- tuple: NEGALIMA keisti ---
koordinates = (54.68, 25.28)
# koordinates[0] = 55.0        # TypeError!

# --- str: NEGALIMA keisti vietoje, sukuriama nauja reikšmė ---
vardas = 'jonas'
print(vardas.capitalize(), vardas)     # originalas nepakito
````

## 2. Kada `if`?

`if` naudojame, kai **sprendimas priklauso nuo sąlygos** ir jį reikia priimti **vieną kartą**.

- `if` — pirmoji sąlyga;
- `elif` — kita sąlyga, jei ankstesnės netiko. Sąlygos tikrinamos **iš viršaus į apačią**,
  todėl laimi **pirmoji teisinga**;
- `else` — visais likusiais atvejais (jau be sąlygos).

**Operatoriai:** `==`, `!=`, `<`, `<=`, `>`, `>=`, `in`, `not in`, `and`, `or`, `not`.

**„Tuščios“ reikšmės laikomos `False`:** `0`, `''`, `[]`, `{}`, `None`.
Todėl rašome `if pazymiai:`, o ne `if len(pazymiai) > 0:`.

> **Dažna klaida:** `=` yra priskyrimas, o `==` yra palyginimas.

````python
# --- if / elif / else: laimi pirmoji teisinga sąlyga ---
pazymys = 7

if pazymys == 10:
    print('Puiku')
elif pazymys >= 8:
    print('Gerai')
elif pazymys >= 5:
    print('Patenkinamai')
else:
    print('Nepatenkinamai')

# --- kelios sąlygos su and / or / not ---
amzius = 20
ar_studentas = True

if amzius >= 18 and ar_studentas:
    print('Studentas su nuolaida')

if amzius < 18 or amzius > 65:
    print('Nuolaida pagal amžių')
else:
    print('Nuolaidos pagal amžių nėra')

# --- narystė: list, set, dict ---
draugai = ['Jonas', 'Rūta', 'Petras']
if 'Rūta' in draugai:
    print('Rūta yra sąraše')

raides = {'a', 'b'}
if 'a' in raides:
    print('a yra aibėje')

studentas = {'vardas': 'Rūta', 'amzius': 20}
if 'amzius' in studentas:
    print('Žodyne yra raktas amzius')
````

## 3. Kada `for`?

`for` naudojame, kai **iš anksto žinome, kiek kartų kartosime**, arba kai reikia
**pereiti per visą kolekciją** (sąrašą, tekstą, žodyną, `range()`).

- `range(n)` → 0, 1, ..., n-1 (**n neįskaitomas**);
- `range(1, n + 1)` → 1, 2, ..., n;
- `range(2, 11, 2)` → 2, 4, 6, 8, 10;
- `enumerate(sarasas, start=1)` → numeris ir reikšmė;
- `zip(a, b)` → dvi eilės vienu metu;
- `.items()` → žodyno raktas ir reikšmė;
- `break` — išeiti iš ciklo, `continue` — praleisti šią iteraciją.

**Kaupimo šablonas**

```python
suma = 0
for x in sarasas:
    suma += x
```

````python
# --- range ---
for i in range(1, 6):
    print(i, end=' ')
print()

# --- per sąrašą su kaupimu ---
pazymiai = [8, 9, 7, 10]
suma = 0
for pazymys in pazymiai:
    suma += pazymys
print(f'Suma: {suma}, vidurkis: {suma / len(pazymiai):.2f}')

# --- per žodyną ---
studentas = {'vardas': 'Rūta', 'amzius': 20}
for raktas, reiksme in studentas.items():
    print(f'{raktas}: {reiksme}')

# --- su numeriu ---
for numeris, pazymys in enumerate(pazymiai, start=1):
    print(f'{numeris}-as pažymys: {pazymys}')

# --- per tekstą ---
for raide in 'labas':
    print(raide, end=' ')
print()

# --- dvi eilės vienu metu ---
vardai = ['Aistė', 'Jonas', 'Petras']
amziai = [20, 21, 19]
for vardas, amzius in zip(vardai, amziai):
    print(f'{vardas} - {amzius} m.')

# --- break ir continue ---
for skaicius in range(1, 11):
    if skaicius == 3:
        continue          # 3 praleidžiame
    if skaicius == 6:
        break             # toliau nebeeiname
    print(skaicius, end=' ')
print()
````

## 4. Kada `while`?

`while` naudojame, kai **nežinome, kiek kartų reikės kartoti** — kartojame tol, kol sąlyga
teisinga.

Trys tipiški atvejai:

1. **Skaičiaus „ardymas“** — kol skaičius > 0;
2. **Augimas iki ribos** — kol reikšmė < 1000;
3. **Įvesties tikrinimas** — `while True:` + `break`, kai sustabdymo sąlyga yra ciklo viduje.

> **Būtina taisyklė:** ciklo viduje turi keistis kintamasis, esantis sąlygoje.
> Pamiršus `+= 1` gaunamas **amžinas ciklas**.

````python
# --- 1) Skaičiaus skaitmenų suma: kiek kartų kartosis, nežinome ---
skaicius = 4729
suma = 0
while skaicius > 0:
    suma += skaicius % 10
    skaicius //= 10    # Ka daro?
print(f'Skaitmenų suma: {suma}')

# --- 2) Augimas iki ribos ---
reiksme = 1
kartai = 0
while reiksme < 1000:
    reiksme *= 2
    kartai += 1
print(f'Prireikė {kartai} kartų, rezultatas {reiksme}')

# --- 3) while True + break (įvesties tikrinimas) ---
# Imituojame vartotojo įvedimą, kad nereikėtų rašyti klaviatūra
bandymai = ['abc', '-5', '12']
bandymas = 0

while True:
    tekstas = bandymai[bandymas]
    bandymas += 1
    try:
        skaicius = int(tekstas)
    except ValueError:
        print(f'{tekstas} - ne skaičius, kartojame')
        continue
    if skaicius > 0:
        break
    print(f'{skaicius} - netinka, kartojame')

print(f'Gautas teigiamas skaičius: {skaicius} (bandymų: {bandymas})')
````

## 5. Kada `def`?

Funkciją rašome, kai:

- tą pačią logiką naudosime **daugiau nei vieną kartą**;
- norime, kad programa būtų **skaitoma** (pavadinimas pasako, ką ji daro);
- norime atskirti **„ką darau“** (funkcijos iškvietimas) nuo **„kaip darau“** (funkcijos kodas).

```python
def funkcijos_vardas(parametras1, parametras2=numatytoji):
    '''Trumpas aprašymas, ką funkcija grąžina.'''
    # kodas
    return rezultatas
```

- `return` **grąžina** reikšmę, kurią galima naudoti toliau;
- `print` tik **atspausdina**, su tokia reikšme toliau dirbti negalima;
- funkcija be `return` grąžina `None`;
- numatytosios reikšmės leidžia kviesti ir `funkcija()`, ir `funkcija(5)`.

````python
def vidurkis(skaiciai):
    '''Grąžina sąrašo vidurkį (0, jei sąrašas tuščias).'''
    if not skaiciai:
        return 0
    return sum(skaiciai) / len(skaiciai)


def sveikinti(vardas='drauge'):
    '''Išveda pasveikinimą.'''
    print(f'Labas, {vardas}!')


def parodyk(suma):
    '''Tik atspausdina - reikšmės grąžinti negalima.'''
    print(f'Suma: {suma}')


def apskaiciuok(a, b):
    '''Grąžina sumą, kurią galima naudoti toliau.'''
    return a + b


# return reikšmę galima naudoti toliau
pazymiai = [8, 9, 7]
print(f'Vidurkis: {vidurkis(pazymiai):.2f}')
print(f'Tuščias sąrašas: {vidurkis([])}')

if vidurkis(pazymiai) >= 8:
    print('Puikus vidurkis')

# numatytoji reikšmė
sveikinti()
sveikinti('Rūta')

# print ir return skirtumas
parodyk(apskaiciuok(2, 3))
print(apskaiciuok(2, 3) * 10)
````

## 6. Kaip išsirinkti? Sprendimų lentelė

| Užduotyje parašyta | Naudokite |
| --- | --- |
| „jei...“, „kai...“, „priklausomai nuo...“ | `if / elif / else` |
| „kiekvienam elementui...“, „pereikite per sąrašą“ | `for` |
| „pakartokite n kartų“, yra `range()` | `for` |
| „kol...“, „kartokite tol, kol...“ | `while` |
| „nežinoma, kiek kartų“ | `while` |
| „parašykite funkciją“, „apskaičiuokite ir grąžinkite“ | `def` |
| „suskaičiuokite, kiek...“ | `for` + skaitiklis arba `sum()` |
| „raskite didžiausią / mažiausią“ | `max()` / `min()` arba `for` + `if` |

Paskutinis pavyzdys — **viskas kartu** vienoje programoje su `main()`.

````python
def ivertinti(pazymys):
    '''Grąžina įvertinimą žodžiu (if sąlygos).'''
    if pazymys >= 9:
        return 'puikiai'
    elif pazymys >= 7:
        return 'gerai'
    elif pazymys >= 5:
        return 'patenkinamai'
    else:
        return 'nepatenkinamai'


def vidurkis(skaiciai):
    '''Grąžina sąrašo vidurkį.'''
    if not skaiciai:
        return 0
    return sum(skaiciai) / len(skaiciai)


def main():
    '''Visa programa: pažymių ataskaita.'''
    pazymiai = [8, 9, 4, 10, 7]

    print('Pažymiai:')
    for numeris, pazymys in enumerate(pazymiai, start=1):
        print(f'  {numeris}. {pazymys} - {ivertinti(pazymys)}')

    print(f'Vidurkis: {vidurkis(pazymiai):.2f}')

    islaike = 0
    for pazymys in pazymiai:
        if pazymys >= 5:
            islaike += 1
    print(f'Išlaikė: {islaike} iš {len(pazymiai)}')

    papildomi = 0
    while vidurkis(pazymiai) < 9:
        pazymiai.append(10)
        papildomi += 1
    print(f'Kad vidurkis būtų 9, reikia {papildomi} papildomų dešimtukų')


main()
````

## 7. Dažniausios klaidos

1. `=` vietoj `==` sąlygoje.
2. `range(1, 5)` duoda `1, 2, 3, 4` — **5 neįtraukiamas**.
3. `while` be `+= 1` → amžinas ciklas.
4. Klaidinga įtrauka (`IndentationError`) arba jos trūkumas po `if` / `for` / `def`.
5. Pamirštas `return` — funkcija grąžina `None`.
6. Sumaišyti `print` ir `return`.
7. `input()` visada grąžina `str` — prieš skaičiuojant reikia `int()` arba `float()`.
8. Sąrašo keitimas ciklo metu (`for` + `remove`) — dažnai praleidžiami elementai.
9. `int('3.5')` → `ValueError`; dešimtainiam skaičiui naudokite `float()`.
10. Dviejų sąlygų sujungimas: `if 5 <= pazymys <= 10:` veikia, o `if pazymys >= 5 or <= 10:`
    yra klaida.
