```python
"""
I užduotis

Kauliukų žaidimas
    Sukurkite programą, kuri meta du žaidimo kauliukus (skaičiai nuo 1 iki 6) ir išveda rezultatą bei jų sumą.
"""
```

```python
# Jūsų kodas
import random

zaidimo_kauliukas1 = random.randint(1, 6)
zaidimo_kauliukas2 = random.randint(1, 6)   
print("Pirmo kauliuko rezultatas:", zaidimo_kauliukas1) 
print("Antro kauliuko rezultatas:", zaidimo_kauliukas2)
suma = zaidimo_kauliukas1 + zaidimo_kauliukas2
print("Kauliukų suma:", suma)   
```

```python
"""
II užduotis

Slaptažodžio generatorius
    Sugeneruokite atsitiktinį 8 simbolių slaptažodį iš didžiųjų ir mažųjų raidžių bei skaičių.
"""
```

```python
# Jūsų kodas
import random
import string

# Generuojame atsitiktinį simbolį   
atsitiktinis_simbolis = random.choice(string.ascii_letters + string.digits)
print("Atsitiktinis simbolis:", atsitiktinis_simbolis)
```

```python
"""
III užduotis

Loterijos numerių generatorius
    Sugeneruokite 5 unikalius loterijos numerius nuo 1 iki 50 (nenaudokite to paties numerio dukart).
"""
```

```python
# Jūsų kodas
import random
simboliai = 'ABCDEFGHIJKLMNOPQRSTUVWXYZabcdefghijklmnopqrstuvwxyz0123456789'
slaptazodis = ''

for i in range(8):
    slaptazodis += random.choice(simboliai) 
print("Atsitiktinis slaptažodis:", slaptazodis)
```

```python
"""
I užduotis

Apskritimo plotas
    Apskaičiuokite apskritimo plotą, kai žinomas spindulys. Naudokite math.pi.
"""
```

```python
"""
II užduotis

Pitagoro teorema
    Apskaičiuokite stačiojo trikampio įžambinę, kai žinomi du statiniai (a=3, b=4).
"""
```

```python
# Jūsų kodas
import math
a = 3
b = 4
izambine = math.sqrt(a**2 + b**2)
print("Stačiojo trikampio įžambinė:", izambine)
```

```python
"""
III užduotis

Kainos suapvalinimas
    Turite kainą 19.23 EUR. Suapvalinkite ją į viršų iki artimiausio euro ir į apačią iki artimiausio euro.
"""
```

```python
# Jūsų kodas
import math
kaina = 19.23
kaina_virsus = math.ceil(kaina)
kaina_apacia = math.floor(kaina)
print("Kaina suapvalinta į viršų:", kaina_virsus)
print("Kaina suapvalinta į apačią:", kaina_apacia)
```

```python
"""
I užduotis

Dienos iki Kalėdų
    Apskaičiuokite, kiek dienų liko iki Kalėdų (gruodžio 25 d.).
"""
```

```python
# Jūsų kodas
import datetime
siandien = datetime.date.today()
kalėdos = datetime.date(siandien.year, 12, 25)
dienos_iki_kaledu = (kalėdos - siandien).days
print("Dienos iki Kalėdų:", dienos_iki_kaledu)
```
