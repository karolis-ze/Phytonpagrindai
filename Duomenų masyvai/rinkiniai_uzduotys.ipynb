```python
# Jūsų kodas
a_set={1, 2, 3, 4, 'a'}
b_set={1, 2, 3, 5, 6, 'b'}

print("Sajunga:", a_set.union(b_set))
print("Bendri elementai:", a_set.intersection(b_set))
print("Skirtumas:", a_set.difference(b_set))
print("Pašalinti bendri elementai:", a_set.symmetric_difference(b_set))

lst = [1,1,2,2,3,4,5,5,5,8,8,1,2]
unikalus_lst = list(set(lst))
print("Unikalus elementai:", unikalus_lst)


```
