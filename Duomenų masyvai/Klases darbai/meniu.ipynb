```python
"""
Užduotis: Kavinės užsakymų sistema
    Sąlyga
        Sukurkite paprastą kavinės užsakymų programą.
            1. Sukurkite meniu žodyną, kuriame būtų patiekalų pavadinimai ir jų kainos. +
            2. Naudodami while ciklą, leiskite vartotojui pasirinkti patiekalus. Įvedus 0, užsakymas baigiamas.
            3. Jei pasirinktas patiekalas egzistuoja meniu, pridėkite jį į užsakymų sąrašą. Jei ne – parodykite klaidos pranešimą.
            4. Naudodami for ciklą, išveskite visus užsakytus patiekalus ir apskaičiuokite bendrą kainą.
            5. Jei bendra suma yra 15 € arba daugiau, pritaikykite 10 % nuolaidą. Kitu atveju nuolaida netaikoma.
            6. Išveskite galutinę mokėtiną sumą.
    """
 
```

```python
meniu = {
    'kava': 20,
    'arbata': 35,
    'kakava': 45,
    'vanduo': 50,
}

uzsakymai = []
while True:
    pasirinkimas = input("Prasome ivesti norima patiekala: (0-iseiti):")

    if pasirinkimas == '0':
        break
    elif pasirinkimas in meniu:
        uzsakymai.append(pasirinkimas)
        print(f"Prideta i uzsakumas {pasirinkimas}")
    else:
        print (f"Tokio pasirinkimo nera")

    
    bendra_suma = 0

    for patiekalas in uzsakymai:
        kaina = meniu[patiekalas]
        print(f'{patiekalas} kurio kaina {kaina}')
        bendra_suma += kaina

    print(f"Bendra uzsakymo suma: {bendra_suma}")

    if bendra_suma >= 15:
        nuolaida = bendra_suma * 0.1
        print(f'Pritaikyta nuolaida {nuolaida}')
    else:
        nuolaida = 0
        print(f'Pritaikyta {nuolaida} proc.')

    print(f'Galutine mokejimo suma {bendra_suma} - {nuolaida}')
```
