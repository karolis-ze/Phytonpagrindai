```python
"""
Skaičiai ir matematinės operacijos

| Funkcija      | Aprašymas             | Python pavyzdys |
| ------------- | --------------------- | --------------- |
|   abs()       | Absoliuti reikšmė     | `abs(-5)` → `5` |
|   bin()       | Dvejetainė forma      | `bin(10)`       |
|   complex()   | Kompleksinis skaičius | `complex(2, 3)` |
|   divmod()    | Dalyba ir liekana     | `divmod(7, 3)`  |
|   float()     | Konvertuoja į float   | `float("3.14")` |
|   hex()       | Šešioliktainė forma   | `hex(255)`      |
|   int()       | Konvertuoja į int     | `int("10")`     |
|   max()       | Didžiausia reikšmė    | `max(1, 5, 3)`  |
|   min()       | Mažiausia reikšmė     | `min([2,1,3])`  |
|   oct()       | Aštuntainė forma      | `oct(8)`        |
|   pow()       | Kėlimas laipsniu      | `pow(2, 3)`     |
|   round()     | Apvalinimas           | `round(3.6)`    |
|   sum()       | Suma                  | `sum([1,2,3])`  |

"""
```

```python
"""
Loginės ir tikrinimo funkcijos

| Funkcija         | Aprašymas                    | Python pavyzdys         |
| ---------------- | ---------------------------- | ----------------------- |
|   all()          | Ar visi `True`               | `all([True, 1])`        |
|   any()          | Ar bent vienas `True`        | `any([0, 2])`           |
|   bool()         | Loginė reikšmė               | `bool(0)`               |
|   callable()     | Ar iškviečiama               | `callable(print)`       |
|   hasattr()      | Ar turi atributą             | `hasattr(obj, "x")`     |
|   isinstance()   | Ar objektas priklauso klasei | `isinstance(5, int)`    |
|   issubclass()   | Ar klasė paveldi             | `issubclass(bool, int)` |

"""
```

```python
"""
Sekos, iteratoriai ir ciklai

| Funkcija        | Aprašymas          | Python pavyzdys                |
| --------------- | ------------------ | ------------------------------ |
|   enumerate()   | Indeksas + reikšmė | `enumerate(['a','b'])`         |
|   filter()      | Filtravimas        | `filter(lambda x:x>0, [-1,2])` |
|   iter()        | Sukuria iteratorių | `iter([1,2,3])`                |
|   len()         | Ilgis              | `len("abc")`                   |
|   map()         | Taiko funkciją     | `map(str, [1,2])`              |
|   next()        | Kitas elementas    | `next(iter([1,2]))`            |
|   range()       | Skaičių seka       | `range(3)`                     |
|   reversed()    | Atvirkštinė seka   | `reversed([1,2])`              |
|   slice()       | Pjūvio objektas    | `slice(1,3)`                   |
|   sorted()      | Rūšiuoja           | `sorted([3,1,2])`              |
|   zip()         | Sujungia sekas     | `zip([1,2],[3,4])`             |

"""
```

```python
"""
Duomenų tipai ir struktūros

| Funkcija         | Aprašymas              | Python pavyzdys        |
| ---------------- | ---------------------- | ---------------------- |
|   dict()         | Žodynas                | `dict(a=1)`            |
|   list()         | Sąrašas                | `list((1,2))`          |
|   set()          | Aibė                   | `set([1,2,2])`         |
|   frozenset()    | Nekintama aibė         | `frozenset([1,2])`     |
|   tuple()        | Tuple                  | `tuple([1,2])`         |
|   bytes()        | Nekintami baitai       | `bytes("abc","utf-8")` |
|   bytearray()    | Keičiami baitai        | `bytearray(b"abc")`    |
|   memoryview()   | Prieiga prie atminties | `memoryview(b"abc")`   |

"""
```

```python
"""
Eilutės ir simboliai

| Funkcija     | Aprašymas            | Python pavyzdys   |
| ------------ | -------------------- | ----------------- |
|   ascii()    | ASCII atvaizdavimas  | `ascii("ž")`      |
|   chr()      | Unicode → simbolis   | `chr(65)`         |
|   ord()      | Simbolis → Unicode   | `ord('A')`        |
|   format()   | Formatavimas         | `format(10, "b")` |
|   repr()     | Oficialus vaizdas    | `repr("a")`       |
|   str()      | Konvertuoja į tekstą | `str(10)`         |

"""
```

```python
"""
Objektai, klasės ir OOP (objekto orientuotas programavimas-object oriented programming)

| Funkcija           | Aprašymas        | Python pavyzdys      |
| ------------------ | ---------------- | -------------------- |
|   object()         | Bazinis objektas | `object()`           |
|   type()           | Tipas / klasė    | `type(5)`            |
|   property()       | Klasės savybė    | `@property`          |
|   classmethod()    | Klasės metodas   | `@classmethod`       |
|   staticmethod()   | Statinis metodas | `@staticmethod`      |
|   super()          | Tėvinė klasė     | `super().__init__()` |

"""
```

```python
"""
Atributai ir aplinka

| Funkcija      | Aprašymas           | Python pavyzdys      |
| ------------- | ------------------- | -------------------- |
|   getattr()   | Gauti atributą      | `getattr(obj,"x")`   |
|   setattr()   | Nustatyti atributą  | `setattr(obj,"x",5)` |
|   delattr()   | Pašalinti atributą  | `delattr(obj,"x")`   |
|   dir()       | Atributų sąrašas    | `dir(str)`           |
|   vars()      | Atributų žodynas    | `vars(obj)`          |
|   globals()   | Globalūs kintamieji | `globals()`          |
|   locals()    | Lokalūs kintamieji  | `locals()`           |
|   id()        | Objekto ID          | `id(x)`              |
|   hash()      | Hash reikšmė        | `hash("abc")`        |

"""
```
