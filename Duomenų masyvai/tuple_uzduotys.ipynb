```python
"""
Užduotis Nr. 1: Koordinačių sistema
    taskai = ((2, 3), (5, 7), (1, 1), (8, 4), (2, 3))
        Suskaičiuokite, kiek kartų pasikartoja taškas (2, 3)
        Raskite, kurioje pozicijoje yra taškas (8, 4)
        Išveskite pirmą ir paskutinį taškus
        Išveskite visus taškus atvirkščia tvarka
        Sukurkite naują tuple su dviem papildomais taškais: (0, 0) - pradžioje ir (10, 10) pabaigoje
"""
```

```python
# Jūsų kodas
taskai = ((2, 3), (5, 7), (1, 1), (8, 4), (2, 3))
print("kordinates 2, 3 pasikartoja:", taskai.count((2, 3)), "kartus")
print("8, 4 pozicija yra:", taskai.index((8, 4)))
print("Pirmas taskas:", taskai[0], "Paskutinis taskas:", taskai[-1])
print("Atvirkscia tvarka:", taskai[-1::-1])
print("Nauja tuple su papildomais taskais:", ((0, 0),) + taskai + ((10, 10),))

```
