```python
"""
Užduotis:
    Paskaičiuoti pelną pardavus visas prekes. Informacija apie kainą ir kiekius yra pateikiama kaip žodynai (dictionary):
    d1 = {"cost_price": 32.67, "sell_price": 45.00, "inventory": 1200}
    d2 = {"cost_price": 225.89, "sell_price": 550.00, "inventory": 100}
    d3 = {"cost_price": 2.77, "sell_price": 7.95, "inventory": 8500}
"""
```

```python
# Jūsų kodas
d1 = {"cost_price": 32.67, "sell_price": 45.00, "inventory": 1200}
d2 = {"cost_price": 225.89, "sell_price": 550.00, "inventory": 100}
d3 = {"cost_price": 2.77, "sell_price": 7.95, "inventory": 8500}

d1_profit = (d1["sell_price"] - d1["cost_price"]) * d1["inventory"]
d2_profit = (d2["sell_price"] - d2["cost_price"]) * d2["inventory"]
d3_profit = (d3["sell_price"] - d3["cost_price"]) * d3["inventory"]
print("Pirmos prekes pelnas:", round(d1_profit, 3), "Antros prekes pelnas:", d2_profit, "Trecios prekes pelnas:", d3_profit)
print("Pelnas pardavus visas prekes:", round(d1_profit + d2_profit + d3_profit, 3))





```
