```python
"""
Taisyklė taikoma visiems žemiau uždaviniams
    eglutės ilgį (eilučių kiekį) turi įvesti vartotojas (input())
    eglutei naudojamas simbolis - *
"""
```

```python
"""
Naudodami for ciklą, išspausdinkite trikampį iš * simbolių (pvz. vartotojas įvedė ilgį 5)

*
**
***
****
*****
"""
```

```python
# Jūsų kodas
ilgis = int(input("Įveskite eglutės ilgį: "))
for i in range(1, ilgis + 1):
    print('*' * i) 
```

```python
"""
Naudodami for ciklą, išspausdinkite apverstą trikampį iš * simbolių (pvz. vartotojas įvedė ilgį 5)

*
**
***
****
*****
"""
```

```python
# Jūsų kodas
ilgis = int(input("Įveskite eglutės ilgį: "))
for i in range(ilgis, 0, -1):  
    print('*' * i)
   
```

```python
"""
Naudodami for ciklą, išspausdinkite trikampį iš * simbolių (pvz. vartotojas įvedė ilgį 5)

    *
   ***
  *****
 *******
*********
"""
```

```python
# Jūsų kodas
ilgis = int(input("Įveskite eglutės ilgį: "))
for i in range(1, ilgis + 2):
    print(' ' * (ilgis - i) + '*' * (2 * i - 1))
```

```python
"""
Naudodami for ciklą, išspausdinkite apverstą trikampį iš * simbolių (pvz. vartotojas įvedė ilgį 5)

*********
 *******
  *****
   ***
    *
"""
```

```python
# Jūsų kodas
ilgis = int(input("Įveskite eglutės ilgį: "))
for i in range(ilgis, 0, -1):
    print(' ' * (ilgis - i) + '*' * (2 * i - 1))
```

```python
"""
Naudodami for ciklą, išspausdinkite tuščiavidurį trikampį iš * simbolių (pvz. vartotojas įvedė ilgį 5)

    *
   * *
  *   *
 *     *
*********
"""
```

```python
# Jūsų kodas
ilgis = int(input("Įveskite eglutės ilgį: "))
for i in range(1, ilgis + 1):
    tarpai = ' ' * (ilgis - i)
    if i == 1:
        print(tarpai + '*')
    elif i == ilgis:
        print('*' * (2 * ilgis - 1))
    else:
        print(tarpai + '*' + ' ' * (2 * i - 3) + '*')   
```
