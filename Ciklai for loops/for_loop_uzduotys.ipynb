```python
"""
Pasinaudodami range sukurkite for loopą, kad gautumėte output su nelyginiais skaičiais:

1
3
5
7
9
"""
```

```python
# Jūsų kodas
for nelyginai_skaiciai in range(10):
    if nelyginai_skaiciai % 2 != 0:
        print(nelyginai_skaiciai)   
```

```python
"""
Pasinaudodami range ir nested for loop, gaukite output:

0 0
0 1
0 2
1 0
1 1
1 2
2 0
2 1
2 2
"""
```

```python
# Jūsų kodas
for i in range(3):
    for j in range(3):
        print(i, j)
```

```python
"""
Pasinaudodami enumerate sukurkite for loopą, kad gautumėte output su indeksu (index) ir verte (value):

0 a
1 b
2 c
"""
```

```python
# Jūsų kodas
lst = ["a", "b", "c"]

for index, value in enumerate(lst):
    print(index, value)
```

```python
"""
Turite žodyną prekes = {"Duona": 1.20, "Pienas": 0.90, "Sūris": 2.50}. Naudodami for ciklą, išspausdinkite tekstą: "Prekė: [pavadinimas], kaina: [kaina]".
"""
```

```python
# Jūsų kodas
dict = {"Duona": 1.20, "Pienas": 0.90, "Sūris": 2.50}
for preke, kaina in dict.items():
    print(f"Prekė: {preke}, kaina: {kaina}")
```

```python
"""
Naudodami for ciklą per skaičių seką nuo 1 iki 10, išspausdinkite visus skaičius, išskyrus 5 (naudokite continue). Jei pasiekiate 8, nutraukite ciklą visiškai (naudokite break).
"""
```

```python
# Jūsų kodas
for skaiciai in range(1, 10):
    if skaiciai == 5:
        continue
    if skaiciai == 8:
        break
    print(skaiciai)
```

```python
"""
Sukurkite sąrašą su keliais kitais sąrašais viduje (pvz., [[1, 2], [3, 4], [5, 6]]). Naudodami ciklą cikle, išspausdinkite kiekvieną skaičių atskiroje eilutėje.
"""
```

```python
# Jūsų kodas
dic = [[1, 2], [3, 4], [5, 6]]
for item in dic:
    for value in item:
        print(value)
```

```python
"""
Pasinaudojant žodynu (dictionaries) sukurkite for loopą, kad gautumėte output:

a 1
b 2
c 3
d 4
e 5
"""
```

```python
# Jūsų kodas
dict = [{"a": 1, "b": 2, "c": 3, "d": 4, "e": 5}]
for item in dict:
    for key, value in item.items():
        print(key, value)
```

```python
"""
Duotas sąrašas:
    miestai = ['Vilnius', 'Kaunas', 'Alytus', 'Rokiškis', 'Ūla', 'Mažeikiai', 'Akmena']
        Spausdinkite tik tuos miestus, kurių pavadinimai ilgesni nei 6 simboliai.
"""
```

```python
# Jūsų kodas
miestai = ['Vilnius', 'Kaunas', 'Alytus', 'Rokiškis', 'Ūla', 'Mažeikiai', 'Akmena']
for miestas in miestai:
    if len(miestas) > 6:
        print(miestas) 
```
