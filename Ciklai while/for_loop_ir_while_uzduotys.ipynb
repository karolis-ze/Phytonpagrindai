```python
"""
Sukurkite programą, kuri spausdina skaičius nuo 1 iki 100, bet:
    - Jei skaičius dalijasi iš 3 IR iš 5 — spausdinkite "FizzBuzz"
    - Jei skaičius dalijasi tik iš 3 — spausdinkite "Fizz"
    - Jei skaičius dalijasi tik iš 5 — spausdinkite "Buzz"
    - Jei skaičius dalijasi iš 7 — prie rezultato pridėkite "!" (pvz. "Fizz!", "14!")
    - Kitu atveju — spausdinkite patį skaičių

Programa veikia cikle: po kiekvieno paleidimo vartotojas gali pasirinkti,
ar nori peržiūrėti rezultatus dar kartą (įvesdamas "taip"), ar baigti (bet kas kita).

Naudokite:
    - while ciklą — kartoti programą pagal vartotojo pageidavimą
    - for ciklą — eiti per skaičius nuo 1 iki 100

Pavyzdys (pirmos kelios eilutės):
    1
    2
    Fizz
    4
    Buzz
    Fizz
    7!
    8
    Fizz
    Buzz
    11
    Fizz
    13
    14!
    FizzBuzz
    ...
    Ar norite paleisti dar kartą? (taip/ne): taip
    (kartojasi iš naujo)
    Ar norite paleisti dar kartą? (taip/ne): ne
    Viso gero!
"""
```

```python
# Jūsų kodas
skaicius = 0
klausimas = "taip"
while klausimas == "taip":
    for skaicius in range(1, 31):
        rezultatas = ""
        if skaicius % 3 == 0 and skaicius % 5 == 0:
            rezultatas = "FizzBuzz"
        elif skaicius % 3 == 0:
            rezultatas = "Fizz"
        elif skaicius % 5 == 0:
            rezultatas = "Buzz"
        else:
            rezultatas = str(skaicius)
        if skaicius % 7 == 0:
            rezultatas += "!"
        print(rezultatas)
    klausimas = input("Ar norite tęsti? (taip/ne): ")
print("Programa baigta. Viso gero!")
```

```python
"""
Vartotojas gali įvesti slaptažodžius tol, kol bus priimtas tinkamas.
Programa leidžia bandyti ne daugiau nei 3 kartus.

Slaptažodis laikomas tinkamu, jei:
    - Ilgis >= 8 simbolių
    - Jame yra bent vienas skaitmuo (0-9)
    - Jame yra bent viena didžioji raidė

Jei slaptažodis netinkamas — pasakykite kodėl (gali būti kelios priežastys vienu metu).
Jei per 3 bandymus netinkamas — spausdinkite "Paskyra užblokuota!"
Jei tinkamas — spausdinkite "Slaptažodis priimtas!"

Naudokite:
    - while ciklą — kartoti bandymus tol, kol liko bandymų ir slaptažodis nepriimtas
    - for ciklą — tikrinti slaptažodžio simbolius (didžiąją raidę ir skaitmenis)

Pavyzdys:
    Įveskite slaptažodį: labas
    - Per trumpas (min. 8 simboliai)
    - Nėra didžiosios raidės
    - Nėra skaičiaus
    Liko bandymų: 2

    Įveskite slaptažodį: Labas123
    Slaptažodis priimtas!
"""
```

```python
# Jūsų kodas
bandymai = 3
teisingas = False
while bandymai > 0 and not teisingas:
    slaptažodis = input("Įveskite slaptažodį: ")
    klaidos = []
    
    if len(slaptažodis) < 8:
        klaidos.append("- Per trumpas (min. 8 simboliai)")
    
    turi_didzioji = False
    turi_skaicius = False
    
    for simbolis in slaptažodis:
        if simbolis.isupper():
            turi_didzioji = True
        if simbolis.isdigit():
            turi_skaicius = True
    
    if not turi_didzioji:
        klaidos.append("- Nėra didžiosios raidės")
    if not turi_skaicius:
        klaidos.append("- Nėra skaičiaus")
    
    if klaidos:
        print("\n".join(klaidos))
        bandymai -= 1
        print(f"Liko bandymų: {bandymai}")
    else:
        teisingas = True
        print("Slaptažodis priimtas!")
    if not teisingas:
        print("Bandymai baigesi." if bandymai == 0 else "")
```

```python
"""
Duotas mokinių pažymių žodynas:

mokiniai = {
    "Jonas": [8, 7, 9, 6, 10],
    "Petras": [5, 4, 6, 5, 7],
    "Ona": [10, 9, 10, 9, 10],
    "Agota": [6, 7, 5, 8, 6],
    "Tomas": [3, 4, 2, 5, 3]
}

Programa leidžia vartotojui ieškoti mokinio rezultatų pagal vardą.
Vartotojas gali ieškoti tiek kartų, kiek nori, kol įves "baigti".

Naudokite:
    - while ciklą — kartoti paiešką tol, kol vartotojas nenori baigti
    - for ciklą — eiti per mokinių žodyną ir pažymių sąrašus

Jei mokinio vardas rastas — išveskite jo vidurkį ir įvertinimą:
    - 9-10: "Puikus"
    - 7-8:  "Geras"
    - 5-6:  "Vidutinis"
    - <5:   "Nepatenkinamas"

Jei mokinio vardas nerastas — pranešite apie tai.
Įvedus "baigti" — programa baigia darbą ir išveda,
kuris mokinys turi aukščiausią vidurkį.

Pavyzdys:
    Įveskite mokinio vardą (arba "baigti"): Jonas
    Jonas: vidurkis 8.0 — Geras

    Įveskite mokinio vardą (arba "baigti"): Petras
    Petras: vidurkis 5.4 — Vidutinis

    Įveskite mokinio vardą (arba "baigti"): Lina
    Mokinys "Lina" nerastas!

    Įveskite mokinio vardą (arba "baigti"): baigti
    Aukščiausias vidurkis: Ona (9.6)
"""
```

```python
# Jūsų kodas
mokiniai = {
    "Jonas": [8, 7, 9, 6, 10],
    "Petras": [5, 4, 6, 5, 7],
    "Ona": [10, 9, 10, 9, 10],
    "Agota": [6, 7, 5, 8, 6],
    "Tomas": [3, 4, 2, 5, 3]
}
paieska = ""

while paieska != "baigti":
    paieska = input('Įveskite mokinio vardą (arba "baigti"): ')
    if paieska == "baigti":
        break
    elif paieska in mokiniai:
        pazymiai = mokiniai[paieska]
        vidurkis = sum(pazymiai) / len(pazymiai)
        if vidurkis >= 9:
            ivertinimas = "Puikus"
        elif vidurkis >= 7:
            ivertinimas = "Geras"
        elif vidurkis >= 5:
            ivertinimas = "Vidutinis"
        else:
            ivertinimas = "Nepatenkinamas"
        print(f"{paieska}: vidurkis {vidurkis:.1f} — {ivertinimas}")
    else:
        print(f'Mokinys "{paieska}" nerastas!')
for vardas in mokiniai:
    if vardas == "Ona":
        auksciausias_vidurkis = sum(mokiniai[vardas]) / len(mokiniai[vardas])
        auksciausias_vardas = vardas
print(f"Aukščiausias vidurkis: {auksciausias_vardas} ({auksciausias_vidurkis:.1f})")
```
