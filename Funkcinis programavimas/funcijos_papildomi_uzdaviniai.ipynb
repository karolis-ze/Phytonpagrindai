```python
"""
Sukurkite funkciją, kuri pagal temperatūrą grąžina tekstą:
    Jei temperatūra < 0 → "Šalta"
    Jei nuo 0 iki 20 (imtinai) → "Vėsu"
    Jei > 20 → "Šilta"
"""
```

```python
# Jūsų kodas
def temperatura(laipsniai):
    if laipsniai < 0:
        return "Salta"
    elif laipsniai >= 0 and laipsniai <= 20:
        return "Vesu"
    else:
        return "Silta"

print(temperatura(-10))    
```

```python
"""
Sukurkite funkciją , kuri patikrina, ar slaptažodis laikomas „stipriu“.
    Slaptažodis laikomas stipriu, jei:
        Jo ilgis yra bent 8 simboliai
        Jame yra bent vienas skaičius
        Funkcija turi grąžinti True arba False.
"""
```

```python
# Jūsų kodas
def slaptazodis(zodis):
    if  len(zodis) < 8:
        return False
    elif not any(char.isdigit() for char in zodis):
        return False
    else:
        return True

print(slaptazodis("slaptazodis"))
```

```python
"""
Sukurkite funkciją, kuri grąžina didžiausią skaičių iš pateikto sąrašo (nenaudojant max()).
    Reikalavimai:
        Sąrašas visada turės bent vieną skaičių
"""
```

```python
# Jūsų kodas
def sarasas_didziausias(sarasas):
    didziausias = sarasas[0]
    for skaicius in sarasas:
        if skaicius > didziausias:
            didziausias = skaicius
    return didziausias

print(sarasas_didziausias([1, 2, 3, 4, 5, 57, 9, 11, 13, 15]))
```

```python
"""
Sukurk funkciją, kuri grąžina antrą didžiausią skaičių iš sąrašo (nenaudojant max() ir sort()).
    Sąlygos:
        Sąraše visada bus bent 2 skaičiai
        Skaičiai gali kartotis
"""
```

```python
# Jūsų kodas
def pav(list):
    pirmas_skaicius = list[0]
    antras_skaicius = list[0]

    for skaicius in list:
        if skaicius > pirmas_skaicius:
            antras_skaicius = pirmas_skaicius
            pirmas_skaicius = pirmas_skaicius
        elif skaicius > antras_skaicius and skaicius != pirmas_skaicius:
            antras_skaicius = skaicius

    return antras_skaicius

print(pav(10, 20, 30, 40))
```

```python
"""
Sukurkite funkciją, kuri grąžina pranešimo aprašą.
    tekstas ir gavejas – position arguments
    svarbus – keyword argument
        Jei svarbus=True, pranešimas pažymimas kaip „SVARBU“
"""
```

```python
# Jūsų kodas
def pranesimo_aprasas(tekstas, gavejas, svarbus=False):
    if svarbus:
        return f"SVARBU: {tekstas} - Gavejas: {gavejas}"
    else:
        return f"{tekstas} - Gavejas: {gavejas}"

print(pranesimo_aprasas("Sveiki", "Jonas", svarbus=True))
```

```python
"""
Sukurkite funkciją, kuri apskaičiuoja galutinę prenumeratos kainą.
    Taisyklės:
        planas (string) gali būti: "basic" (10 €/mėn), "pro" (20 €/mėn), "premium" (30 €/mėn)
        menesiai – prenumeratos trukmė (position argument)
        nuolaida_procentais ir mokestis – keyword arguments
            Nuolaida taikoma visai sumai procentais
            Galiausiai pridedamas papildomas mokestis
"""
```

```python
# Jūsų kodas
def prenumeratos_kaina(planas, menesiai, nuolaida_procentais=0, mokestis=0):
    if planas == "basic":
        menesio_kaina = 10
    elif planas == "pro":
        menesio_kaina = 20
    elif planas == "premium":
        menesio_kaina = 30
    else:
        return "Neteisingas planas"

    bendra_kaina = menesio_kaina * menesiai

    bendra_kaina -= bendra_kaina * (nuolaida_procentais / 100)
    bendra_kaina += mokestis

    return f"Prenumeratos kaina su {planas} planu {menesiai} mėnesių: {bendra_kaina:.2f} €"

print(prenumeratos_kaina("pro", 6, nuolaida_procentais=10, mokestis=5))
```

```python
"""
Sukurkite funkciją, kuri apskaičiuoja egzamino rezultatą procentais ir grąžina įvertinimą.
    Taisyklės:
        taskai ir max_taskai – position arguments
        islaikyta_riba ir apvalinti – keyword arguments
        Procentai = (taskai / max_taskai) * 100
        Jei apvalinti=True, procentai apvalinami iki sveiko skaičiaus (round)
        Jei procentai ≥ islaikyta_riba → "Išlaikyta", kitu atveju "Neišlaikyta"
"""
```

```python
# Jūsų kodas
def egzo_rezultatas(taskai, max_taskai, islaikyta_riba=0, apvalinti=True):
    procentai = (taskai / max_taskai) * 100
    if apvalinti:
        procentai = round(procentai)
    if procentai >= islaikyta_riba: 
        ivertinimas = "Išlaikyta"
    else:
        ivertinimas = "Neišlaikyta"    

    return f"{ivertinimas} ({procentai}%)"

print(egzo_rezultatas(80, 95, islaikyta_riba=25, apvalinti=True))
   
    
```

```python
"""
Sukurkite funkciją, kuri apskaičiuoja siuntimo kainą.
    Taisyklės:
        svoris (kg) ir atstumas (km) – position arguments
        express ir draudimas – keyword arguments
        Bazinė kaina: svoris * 2 + atstumas * 0.1
        Jei express=True, kaina padidinama 50 %
        Jei draudimas=True, pridedami 5 €
"""
```

```python
# Jūsų kodas
def siuntimo_kaina(svoris, atstumas, express=True, draudimas=True):
    bazine_kaina = svoris * 2 + atstumas * 0.1
    if express:
        bazine_kaina += bazine_kaina * 0.5
    if draudimas:
        bazine_kaina += 5 
    return bazine_kaina

print(siuntimo_kaina(50, 300, express=False, draudimas=False))
```
