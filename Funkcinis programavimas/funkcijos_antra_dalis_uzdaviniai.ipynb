```python
"""
1 užduotis

Sukurkite funkciją ar_teigiamas(skaicius), kurios viduje būtų kita funkcija tikrinti().
    Vidinė funkcija turi patikrinti, ar skaičius didesnis už 0.
"""
```

```python
# Jūsų kodas
def ar_teigiamas(a):

    def tikrinimas(a):
        if a > 0:
            print(f"Skaicius {a} didesnis uz 0")
        else:
            print(f"Skaicius {a} mazesnis arba lygus nuliui")

    tikrinimas(a)

ar_teigiamas(-5)
```

```python
"""
2 užduotis

Sukurkite funkciją galutine_kaina(kaina), kuri apskaičiuoja galutinę kainą su PVM.
    PVM skaičiavimas turi būti realizuotas vidinėje funkcijoje, kurios negalima pasiekti iš išorės.

Taisyklės:
    PVM = 21 %
    Vidinė funkcija grąžina tik PVM sumą
"""
```

```python
# Jūsų kodas
def galutine_kaina(a):

    def kaina_su_pvm(a):
        kaina = a * 1.21
        print(f"Kaina su PVM nuo kainos {a} yra {kaina} Euru")

    kaina_su_pvm(a)

galutine_kaina(15)
```

```python
"""
3 užduotis

Sukurkite funkciją tikrinti_slaptazodi(slaptazodis), kuri tikrina, ar slaptažodis tinkamas
    Naudokite dvi vidines funkcijas, kurių kiekviena atlieka atskirą patikrą:
        ar_ilgas() – ar ilgis bent 8 simboliai
        turi_skaiciu() – ar turi bent vieną skaičių
"""
```

```python
# Jūsų kodas
def slaptazodis(zodis):

    def tikrinimas(zodis):
        if len(zodis) < 8:
            return "pateiktas zodis pertrumpas"
        elif not any(char.isdigit() for char in zodis):
            return "Zodyje nera skaiciu"
        else:
            return f"Pateiktas zodis {zodis} tinkamas"
    
    return tikrinimas(zodis)
print(slaptazodis("manovardas"))
```

```python
"""
4 užduotis

Duotas skaičių sąrašas
    Naudojant map() ir atskirą funkciją, sukurk naują sąrašą, kuriame visi skaičiai pakelti kvadratu.
"""
```

```python
skaiciai = [1, 2, 3, 4, 5]

rezultatas = map(lambda x: x ** 2, skaiciai)

print(list(rezultatas))
```

```python
"""
5 užduotis

Duotas vardų sąrašas
    Naudojant map(), pakeisk vardus taip, kad pirmoji raidė būtų didžioji
"""
```

```python
# Jūsų kodas
vardai = ["petras", "antanas", "jonas", "kipras"]

rezultatas = list(map(str.title, vardai))

print(rezultatas)
```

```python
"""
6 užduotis

Duotas kainų sąrašas be PVM
    Naudojant map() ir atskirą funkciją, apskaičiuok kainas su 21 % PVM
"""
```

```python
# Jūsų kodas
kainos = [15, 20, 35, 45, 55, 70]

rezultatas = list(map(lambda x: round(x * 1.21, 2), kainos))
print((rezultatas))
```

```python
"""
7 užduotis

Duotas skaičių sąrašas
    Naudojant filter() ir lambda, palik tik lyginis skaičius
"""
```

```python
# Jūsų kodas
skaiciai =  [15, 20, 35, 45, 55, 70]
lyginiai_skaiciai = list(filter(lambda x: x % 2 == 0, skaiciai))

print(lyginiai_skaiciai)
```

```python
"""
8 užduotis

Duotas kainų sąrašas
    Naudojant map() ir lambda:
        kainas < 20 padidink 10 %
        kainas ≥ 20 palik nepakitusias
"""
```

```python
# Jūsų kodas
kainu_sarasas = [15, 20, 30, 40, 99]

naujos_kainos = list(map(lambda x: x * 1.1 if x < 20 else x, kainu_sarasas))

print(naujos_kainos)
```
