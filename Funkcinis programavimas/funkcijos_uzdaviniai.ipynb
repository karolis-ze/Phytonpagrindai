```python
"""
Parašykite programą, kuri sukurtų funkciją func(), kuri priimančią kintamą skaičių argumentų ir atspausdinančią kiekvieną jų reikšmę

Pastaba: sukurkite šią funkciją taip, kad ji galėtų priimti bet kokį skaičių argumentų, juos apdoroti ir parodyti kiekvieno argumento reikšmę.
"""
```

```python
# Jūsų kodas
def func(*args):
    for arg in args:
        print(f"{arg}")  
        
func(1, 2, 3, 4, 5)       
```

```python
"""
Parašykite funkciją calculation(), kuri:
    priima du kintamuosius ir apskaičiuoja jų sudėtį bei atimtį
    funkcija turėtų grąžinti ir sumą, ir skirtumą vienu return sakiniu
"""
```

```python
# Jūsų kodas
def calculation(a, b):
    suma = a + b
    skirtumas = a - b
    return suma, skirtumas

calculation(10, 5)
```

```python
"""
Parašykite funkciją, kuri rastų didžiausią iš dviejų skaičių
"""
```

```python
# Jūsų kodas
def didziausias(a,b):
    if a > b:
        return f"Didziausias skaicius {a}"
    elif b > a:
        return f"Didziausias skaicius {b}"
    else:
        return f"Abu skaiciai {a} ir {b} yra lygūs"

didziausias (50, 50)
```

```python
"""
Parašyti funkciją greičiui tikrinti:
    Funkcija paima argumentą - automobilio greitį
        Jei greitis 50 ar mažesnis, funkcija grąžina ”Ok”,
        jei greitis didesnis nei 50 - už kiekvieną 5km/h greičio viršijimą duodamas baudos taškas (pvz: greitis 70, atsakymas - 4)
        Jei surenkami 8 taškai ir daugiau - funkcija ne tik grąžina taškų sumą, bet ir priduria, jog vairuotojo teisės atimamos.
"""
```

```python
# Jūsų kodas
def autogreitis(greitis):
    if greitis <= 50:
        return "Ok"
    else:
        baudos_taskai = (greitis - 50) // 5
        if baudos_taskai >= 8:
            return f"Baudos taškai: {baudos_taskai}. Vairuotojo teisės atimamos."
        else:
            return f"Baudos taškai: {baudos_taskai}"

autogreitis(150)
```

```python
"""
Parašykite funkciją, kuri:
    paima eilutę kaip įvestį ir grąžina žodyną, kuriame yra didžiųjų ir mažųjų raidžių skaičius eilutėje
    visi simboliai, kurių negalima priskirti didžiosioms ar mažosioms raidėms (pvz., simboliai), turėtų būti laikomi „kitais“
"""
```

```python
# Jūsų kodas
def zodynas(eilute):
    didziosios = sum(1 for char in eilute if char.isupper())
    mazosios = sum(1 for char in eilute if char.islower())
    kiti = len(eilute) - didziosios - mazosios
    return {"Didžiosios": didziosios, "Mažosios": mazosios, "Kiti": kiti}

zodynas("Karolina123!@#")
```

```python
"""
Parašykite funkciją, kuri:
    kaip įvestį priima sąrašą ir elementą
    funkcija turėtų pridėti elementą prie sąrašo tik tuo atveju, jei jo jame dar nėra
    panaudokite not in
"""
```

```python
# Jūsų kodas
def elementas_sarase(sarasas, elementas):
    if elementas not in sarasas:
        sarasas.append(elementas)
    else:
        print(f"Elementas {elementas} jau yra sąraše.")
    return sarasas  

elementas_sarase([1, 2, 3], 2)
```

```python
"""
Parašykite funkciją, kuri:
    kaip įvestį paima skaičių sąrašą
    grąžina surūšiuotą sąrašą, kuriame yra tik teigiami skaičiai iš įvesties sąrašo
    naudokite append, sorted
"""
```

```python
# Jūsų kodas
def elementas_sarase(sarasas):
    teigiami_skaiciai = []
    for skaicius in sarasas:
        if skaicius > 0:
            teigiami_skaiciai.append(skaicius)
    return sorted(teigiami_skaiciai)

elementas_sarase([-1, 2, -3, 4, 0, 5])
```
