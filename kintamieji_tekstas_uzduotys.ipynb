```python
"""
V užduotis:
    Pasirinkti norimą žinių portalą:
        Iš antraštės atrinkite:
            pirmus 10 simbolių;
            paskutinius 10;
            kas antrą;
            iškirpkite tekstą nuo 10-to iki 20-to simbolių;
"""
```

```python
# Jūsų kodas
antraste = "Astronomai tolimoje galaktikoje aptiko paslaptingą energijos žybsnį"
pirmi_10 = antraste[:10]
paskutiniai_10 = antraste[-10:]
kas_antra = antraste[::2]
nuo_10iki20 = antraste[10:20]
print("Pirmi 10 simboliu:", pirmi_10)
print("Paskutiniai 10 simbolius:", paskutiniai_10)
print("Kas antras simbolis:", kas_antra)
print("Nuo 10 iki 20 simboliu:", nuo_10iki20)
```
