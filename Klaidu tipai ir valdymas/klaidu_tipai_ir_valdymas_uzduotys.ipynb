```python
"""
14. „Amžiaus patikra“ Užduotis.
    Paprašykite vartotojo įvesti amžių. Jei amžius mažesnis už nulį, naudodami raise sukelkite ValueError. Klaida turi būti apdorota naudojant try/except
"""
```

```python
# Jūsų kodas
ivestis = input("Iveskite savo amziu: ")

try:
    amzius = int(ivestis)
    if amzius < 0:
        raise ValueError("Amzius negali buti neigiamas!")
except ValueError as e:
    print(f"Klaida: {e}")

else:
    print(f"Jusu amzius: {amzius}")

finally:
    print("Amziaus patikra baigta.")
```

```python
"""
15. „Teigiamo skaičiaus tikrinimas“ Užduotis.
    Sukurkite funkciją tikrinti_skaiciu(skaicius), kuri sukelia ValueError, jei skaičius nėra teigiamas. Jei skaičius tinkamas, funkcija jį grąžina
"""
```

```python
# Jūsų kodas
def tikrinti_skaiciu(skaicius):
    if skaicius <= 0:
        raise ValueError("Skaicius turi buti teigiamas!")
    return skaicius 
try:
    rezultatas = tikrinti_skaiciu(-5)
    print(f"Skaicius: {rezultatas}")
except ValueError as e:
    print(f"Klaida: {e}")
finally:
    print("Skaiciaus patikra baigta.")
```

```python
"""
16. „Slaptažodžio ilgis“ Užduotis.
    Sukurkite funkciją, kuri patikrina slaptažodį. Jei jo ilgis mažesnis nei 8 simboliai, naudokite raise ValueError. Kitu atveju išveskite, kad slaptažodis tinkamas
"""
```

```python
# Jūsų kodas
def slaptazodis(zodis):
    if len(zodis) >= 8:
        return zodis
    else:
        raise ValueError("Per mažai simbolių! Slaptažodis turi būti bent 8 simbolių ilgio.")
    
try:
    rezultatas = slaptazodis("vilniusgo")
    print(f"Pasirinktas {rezultatas} tinkamas")

except ValueError as klaida:
    print(f"Klaida: {klaida}")

except NameError as klaida:
    print(f"Klaida: {klaida}")
          
finally:
    print("Patikra baigta!")
```

```python
"""
17. „Banko sąskaitos likutis“ Užduotis.
    Sukurkite funkciją nuskaiciuoti_likuti(likutis, suma). Ji turi sukelti ValueError, jei nuskaitoma suma yra neigiama arba didesnė už turimą likutį. Jei sąlygos tenkinamos, grąžinkite naują likutį
"""
```

```python
# Jūsų kodas
def nuskaiciuoti_likuti(likutis, suma):
    if suma > likutis:
        raise ValueError("Nepakankamas likutis")
    if suma < 0:
        raise ValueError("Suma turi buti teigiama!")
    
    likutis -= suma
    return likutis

try:
    rezultatas = nuskaiciuoti_likuti(50, -55)
    print(f"Likutis atskaicius suma yra: {rezultatas}")

except ValueError as klaida:
    print(f"Klaida: {klaida}")

finally:
    print("Patikra baigta")
```

```python
"""
18. „Temperatūros konverteris“ Užduotis.
    Sukurkite funkciją, kuri Celsijaus temperatūrą konvertuoja į Farenheito temperatūrą. Jei temperatūra mažesnė už absoliutų nulį, sukelkite ValueError. Klaida turi būti apdorota programoje
"""
```

```python
# Jūsų kodas
def temperatura_celsijus_i_farenheita(celsijus):
    if celsijus < -273.15:
        raise ValueError("Temperatura negali buti zemesne uz absoliutu nuli!")
    farenheitas = (celsijus * 9/5) + 32
    return farenheitas

try:
    pavertimas = temperatura_celsijus_i_farenheita(celsijus=30)
    print(f"{celsijus} laipsniu pagal Celciju atitinka {pavertimas} pagal Farenheita")

except ValueError as klaida:
    print(f"Klaida: {klaida}")

finally:
    print("Patikrinimas baigtas!")
```

```python
"""
Papildoma užduotis: „Vienas saugus duomenų apdorojimo procesas“ Užduotis

Sukurkite programą, kuri:
    1. Paprašo vartotojo įvesti du skaičius
    2. Apskaičiuoja pirmojo skaičiaus dalmenį iš antrojo
    3. Jei įvestis netinkama, apdoroja ValueError
    4. Jei antrasis skaičius lygus nuliui, apdoroja ZeroDivisionError
    5. Jei rezultatas neigiamas, funkcija turi sukelti ValueError naudodama raise
    6. Naudoja else sėkmingam rezultatui išvesti ir finally programos pabaigos pranešimui
"""
```

```python
# Jūsų kodas
skaicius_1 = input("Iveskite skaiciu:")
skaicius_2 = input("Iveskite antra skaicius:")

try:
    dalyba = int(skaicius_1) / int(skaicius_2)
    if int(skaicius_1) / int(skaicius_2) < 0:
        raise ValueError("Rezultatas negali buti neigiamas!")    

except ValueError as klaida:
    print(f"Ivedete netinkama skaicius {klaida}")

except ZeroDivisionError:
    print("Skaiciaus negalima dalinti is 0")

else:
    print(f"Pirma padalinus is antro gausime: {dalyba}")

finally:
    print("Tikrinimas baigtas!")
```
